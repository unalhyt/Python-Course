## 1. Printing Mixed Data Types
The `print()` function is used to display output. When you want to combine text (strings) and numbers in a single print statement using concatenation (`+`), you must convert the numbers to strings first using `str()`.

**Example:**

```python
x = 3
y = 5

# This would cause an error because you cannot add integers to strings directly:
# print(x + ' times ' + y + ' equals ' + x*y)

# Correct way: Convert numbers to strings first
print(str(x) + ' times ' + str(y) + ' equals ' + str(x*y))
```

### Controlling Newlines
By default, `print()` moves to the next line after printing. You can prevent this by setting the `end` parameter to an empty string or a space.


```python
print('Hello', end=' ')
print('World!')
```

## 2. Getting User Input
To get information from the user, use the `input()` function. You can provide a prompt string inside the parentheses.

**Important:** The `input()` function **always** returns the user's input as a **string**, even if they type a number.


```python
# input('Feed me a stray cat.') # The output of this function will be a string even if you enter a number.

ans = input('Feed me a stray cat: ')

print("You entered:", ans)

# Check the type of the variable
print("Type of input:", type(ans))
```

## 3. Converting Input to Numbers
Since `input()` returns a string, if you want to perform mathematical operations on the input, you must convert (cast) it to a number type like `int` or `float`.


```python
# Convert the string input to a float
# Note: This will fail if 'ans' contains non-numeric characters
try:
    ans_num = float(ans)
    print("Converted value:", ans_num)
    print("Type after conversion:", type(ans_num))
except ValueError:
    print("Could not convert input to a number.")
```

## 4. Practice Task: Pythagorean Theorem
**Exercise:**
1.  Ask the user for the lengths of two sides of a right triangle (`a` and `b`).
2.  Compute the length of the hypotenuse (`c`) using the formula: $a^2 + b^2 = c^2$ (so $c = \sqrt{a^2 + b^2}$).
3.  Print the result.


```python
# Get two numbers from the user
side1 = float(input('Length of side a: '))
side2 = float(input('Length of side b: '))

# Compute the length of side c
# Note: raising to the power of 0.5 (or 1/2) is the same as taking the square root
side3 = (side1**2 + side2**2)**(1/2)

print(' ')
print('The length of side c is ' + str(side3))
```
