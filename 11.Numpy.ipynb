## Python Basics — NumPy

In this notebook, we will explore **NumPy**, a fundamental library for scientific computing in Python. NumPy provides support for large, multi-dimensional arrays and matrices, along with a collection of high-level mathematical functions to operate on these arrays.

## Learning Goals
By the end of this notebook, students will be able to:

- Use NumPy functions like `np.mean()`, `np.linspace()`, and `np.sort()`.

## 1. Importing NumPy
To use NumPy, you first need to import it. It is standard practice to alias it as `np`.

**Note:** If you restart the runtime, you must re-import the library.


```python
import numpy as np

# Create a list for demonstration
lst = [10, 20, 30]

# Calculate the mean using NumPy
mean_val = np.mean(lst)
print("Mean of lst:", mean_val)
```

## 2. Creating Arrays
NumPy's main object is the homogeneous multidimensional array. You can create an array from a regular Python list.


```python
number = [1, 2, 3, 4, 5, 6]
print("Original list type:", type(number))

# Convert list to NumPy array
numbernp = np.array(number)
print("NumPy array type:", type(numbernp))

print("List:", number)
print("NumPy Array:", numbernp)
```

## 3. Array Functions
NumPy provides many useful functions for generating data. For example, `np.linspace()` creates an array of evenly spaced numbers over a specified interval.


```python
# Create 20 linearly spaced numbers from 1 to 8
linspace_array = np.linspace(1, 8, 20)
print("Linspace array:", linspace_array)
```

## 4. Array Methods
NumPy arrays have built-in methods for common operations like finding the minimum or maximum value.


```python
print("Minimum value:", numbernp.min())
print("Maximum value:", np.max(numbernp))
```

## 5. Sorting: Method vs. Function
It's important to understand the difference between the `.sort()` method and the `np.sort()` function.

- `list.sort()` modifies the list **in-place** (changes the original list).
- `np.sort(array)` returns a **new sorted copy** of the array (does not change the original).

**Example:**

```python
# Using list.sort() - Modifies in-place
n = [4, 5, 6, 7, 1, 2, 6, 4, 8]
print("Original list n:", n)
n.sort()
print("Sorted list n (in-place):", n)

# Using np.sort() - Returns a new array
n = [4, 5, 6, 7, 1, 2, 6, 4, 8]
print("Original list n (reset):", n)
print("Type before np.sort:", type(n))

sorted_n = np.sort(n)
print("Sorted array (new copy):", sorted_n)
print("Original list n (unchanged):", n)
print("Type of sorted_n:", type(sorted_n))
```

## 7. Why NumPy? (Lists vs Arrays)

You already learned about Python **lists** in **Notebook 4**. So why do we need a new "array" type?

**Analogy:** A Python list is like a backpack — flexible, can hold anything (numbers, strings, other lists). A NumPy array is like a **tool-box drawer for numbers only** — fixed-type, compact, and built for fast math.

| Feature | Python list | NumPy array |
|---|---|---|
| Element types | Mixed | All the same type |
| Math on whole thing | Requires a loop | Works directly: `a * 2` |
| Memory usage | Higher | Lower |
| Speed for big data | Slow | Very fast (vectorized) |
| Shape/dimensions | 1D only (nesting required) | Native 1D, 2D, 3D, … |

Let's see the speed difference on a million numbers.

```python
import time

# Square one million numbers with a Python list (loop)
py_list = list(range(1_000_000))
t0 = time.time()
py_squared = [x * x for x in py_list]
t_list = time.time() - t0

# Square one million numbers with a NumPy array (vectorized)
np_arr = np.arange(1_000_000)
t0 = time.time()
np_squared = np_arr ** 2
t_np = time.time() - t0

print(f"Python list took: {t_list:.4f} s")
print(f"NumPy array took: {t_np:.4f} s")
print("NumPy is roughly", round(t_list / max(t_np, 1e-9), 1), "x faster")
```

## 8. More Ways to Create Arrays

Typing a list by hand is fine for small arrays, but often you need many elements that follow a pattern. NumPy has shortcut functions for that:

- `np.zeros(n)` — array of `n` zeros.
- `np.ones(n)` — array of `n` ones.
- `np.full(n, value)` — array of `n` copies of `value`.
- `np.arange(start, stop, step)` — like `range()` but returns an array; `stop` is **not** included.
- `np.linspace(start, stop, count)` — `count` evenly spaced numbers; `stop` **is** included.
- `np.eye(n)` — `n × n` identity matrix.

**`arange` vs `linspace`:** use `arange` when you know the **step**, use `linspace` when you know how **many points** you want.

```python
# Different ways to create arrays without typing every element
print("zeros:  ", np.zeros(5))           # five 0.0 values
print("ones:   ", np.ones(4))            # four 1.0 values
print("full:   ", np.full(3, 7.5))       # repeat 7.5 three times
print("arange: ", np.arange(0, 10, 2))   # start, stop (exclusive), step
print("linspace:", np.linspace(0, 1, 5)) # start, stop (inclusive), count
print("eye:\n", np.eye(3))               # 3x3 identity matrix
```

## 9. Array Attributes

Every NumPy array carries information about itself. These attributes are useful for checking that your data has the shape you expect.

- `.shape` — the size along each dimension (like `(rows, columns)` for 2D).
- `.ndim` — number of dimensions.
- `.size` — total number of elements.
- `.dtype` — the data type of the elements (e.g. `int64`, `float64`).

```python
# Inspect an array's "identity card"
sensor = np.array([1.5, 2.3, 3.1, 4.8, 5.0])

print("shape:", sensor.shape)     # number of elements in each dimension
print("ndim: ", sensor.ndim)      # number of dimensions (1 = list-like)
print("size: ", sensor.size)      # total number of elements
print("dtype:", sensor.dtype)     # type of the elements (e.g. float64)
```

## 10. Element-wise Math & Vectorization

With normal Python lists you must loop to add or multiply numbers. With NumPy arrays, math operators work on **every element at once**. This is called **vectorization** and it's one of the main reasons NumPy is so fast.

Think of it like baking cookies: a regular list forces you to bake one at a time, while a NumPy array lets you bake the whole tray in a single oven cycle.

```python
# Element-wise arithmetic: math is applied to every element at once
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("a + b  =", a + b)       # [11 22 33 44]
print("a * b  =", a * b)       # [10 40 90 160]
print("a ** 2 =", a ** 2)      # [1 4 9 16]

# NumPy math functions also work element-wise
angles = np.array([0, 30, 60, 90])           # degrees
print("sin:", np.sin(np.deg2rad(angles)))    # convert to radians first
print("sqrt:", np.sqrt(np.array([1, 4, 9, 16])))
```

## 11. Useful Statistics

NumPy gives you fast one-call summaries of data. These are the functions you'll reach for almost every day when working with experimental data.

| Function | What it returns |
|---|---|
| `np.sum(a)` | Sum of all elements |
| `np.mean(a)` | Average |
| `np.median(a)` | Middle value |
| `np.std(a)` | Standard deviation |
| `np.argmin(a)` / `np.argmax(a)` | Index of smallest / largest value |

```python
# Pretend these are sensor readings from an experiment
data = np.array([4.2, 5.1, 4.8, 5.5, 4.9, 5.0, 4.7, 5.3])

print("Sum:    ", np.sum(data))        # total
print("Mean:   ", np.mean(data))       # average
print("Median: ", np.median(data))     # middle value
print("Std:    ", np.std(data))        # standard deviation (spread)
print("Min idx:", np.argmin(data))     # index of smallest value
print("Max idx:", np.argmax(data))     # index of largest value
```

## 13. Engineering Example 1 — Ohm's Law (Electrical Engineering)

A common **electrical engineering** task: given a fixed voltage, compute the current through a list of different resistors using Ohm's law:

`I = V / R`

With plain Python you'd write a loop. With NumPy, one line handles the whole list — this is the power of **vectorization** in action.

```python
# Ohm's Law: I = V / R  for many resistor values at once
voltage = 12.0                                  # fixed voltage in volts
resistors = np.array([100, 220, 470, 1000, 2200, 4700])   # ohms

current = voltage / resistors                   # element-wise division
power = voltage * current                       # P = V * I, also element-wise

print("Resistor (Ohm):", resistors)
print("Current (A):   ", np.round(current, 4))
print("Power (W):     ", np.round(power, 4))
print("Highest current at R =", resistors[np.argmax(current)], "Ohm")
```

## 14. Engineering Example 2 — Projectile Motion

A classic **mechanical / physics engineering** problem: a ball is launched with initial speed `v0` at angle `θ`. Compute its position at many time steps.

Equations (ignoring air drag):

- `x(t) = v0 · cos(θ) · t`
- `y(t) = v0 · sin(θ) · t − 0.5 · g · t²`

With NumPy, you compute **x and y for all time steps at once** using one array expression — no loop needed.

```python
# Projectile motion trajectory using vectorized time array
# Mechanical engineering / physics classic problem.

v0 = 20.0                       # initial speed in m/s
angle_deg = 45.0                # launch angle in degrees
g = 9.81                        # gravitational acceleration in m/s^2

angle_rad = np.deg2rad(angle_deg)             # convert degrees to radians
v0x = v0 * np.cos(angle_rad)                  # horizontal velocity component
v0y = v0 * np.sin(angle_rad)                  # vertical velocity component

t = np.linspace(0, 3, 7)                      # 7 time samples from 0 to 3 s
x = v0x * t                                   # horizontal position at each t
y = v0y * t - 0.5 * g * t**2                  # vertical position at each t

print("Time (s): ", np.round(t, 2))
print("x (m):    ", np.round(x, 2))
print("y (m):    ", np.round(y, 2))
print("Max height reached:", round(y.max(), 2), "m")
```

## 15. Common Pitfalls

1. **Forgetting the list brackets.** `np.array(1, 2, 3)` raises an error. You must pass a list: `np.array([1, 2, 3])`.
2. **Integer dtype truncation.** If you create `a = np.array([1, 2, 3])` (all ints) and write `a[0] = 1.7`, NumPy stores `1`, not `1.7`. Use floats: `np.array([1.0, 2.0, 3.0])`.
3. **`arange` with floats.** `np.arange(0, 1, 0.1)` may give 10 or 11 elements due to floating-point rounding. Prefer `np.linspace(0, 1, 11)` when the endpoint matters.
4. **`.sort()` vs `np.sort()`.** The first modifies in place and returns `None`. The second returns a new array. Don't write `b = a.sort()` — `b` will be `None`.
5. **Missing import.** If you see `NameError: name 'np' is not defined`, you forgot `import numpy as np` (or restarted the kernel).
6. **Shape mismatch in math.** Adding arrays of different sizes raises an error unless shapes are compatible (see Notebook 25 on broadcasting).
