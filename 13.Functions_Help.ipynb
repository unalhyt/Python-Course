## 1. Using `help()`
The `help()` function is the most standard way to access documentation in Python. It prints the docstring of the object you pass to it. Think of it like a built-in manual — whenever you are unsure what a function does, `help()` will show you its description, what arguments it takes, and what it returns.

**Example:**

````python
alist = [1, 2, 3]

# Get help on the sum function
help(sum)

# Example usage based on help:
# sum(iterable, start=0)
print("Sum with start=10:", sum(alist, 10))
````

You can also use `help()` on **methods** — functions that belong to an object. This is useful when you want to know what a specific method does, for example the `.append()` method on a list (we learned about lists in Notebook 4).

````python
# Get help on a specific method of a type
help(list.append)
````

### How to Read a Function Signature

When you call `help()`, you will often see something like this:

```
sum(iterable, /, start=0)
```

Here is what each part means:
- **`sum`** — the name of the function
- **`iterable`** — a required argument (the data you pass in, like a list)
- **`start=0`** — an optional argument with a default value of `0`
- **`/`** — means the arguments before it are "positional only" (you cannot use `start=...` style for them)

Think of a function signature like a recipe title — it tells you what ingredients (arguments) you need and which ones are optional.

## 2.5 Using `dir()` to Explore Objects

The `dir()` function lists all the attributes and methods available on an object. Think of it like opening a toolbox — `dir()` shows you every tool inside, even if you do not know what each one does yet. You can then use `help()` on any item you find interesting.

**Example:**

````python
# List all methods available on a list object
my_list = [1, 2, 3]

# Filter out private methods (those starting with _)
public_methods = [m for m in dir(my_list) if not m.startswith('_')]
print("Public methods on a list:")
print(public_methods)
````

````python
import numpy as np

# You can also use dir() on imported libraries
# This shows what functions NumPy offers (first 15)
numpy_items = [m for m in dir(np) if not m.startswith('_')]
print("First 15 public items in NumPy:")
print(numpy_items[:15])
````

## 2.6 Tab Completion in Jupyter

One of the fastest ways to explore an object is **tab completion**. In a Jupyter notebook or IPython terminal:

1. Type a variable name followed by a dot: `my_list.`
2. Press the **Tab** key
3. A dropdown menu will appear showing all available methods and attributes

This is a quick alternative to `dir()` — you can browse methods visually and select one. Once you find a method, use `help()` or `?` to learn more about it.

> **Tip:** Tab completion also works for file paths, module names, and variable names. It saves a lot of typing!

## 3. Inspecting Source Code
For more advanced users, the `inspect` module allows you to retrieve the source code of a function or class. This is useful for understanding how a library function is implemented.


````python
import numpy as np
import inspect

# Get the source lines of np.linspace
source_lines = inspect.getsourcelines(np.linspace)
print("Source code for np.linspace (first 5 lines):")
for line in source_lines[0][:5]:
    print(line.strip())
````

## 4. Troubleshooting Errors
When you encounter an error, Python shows a **traceback** — a message that tells you exactly what went wrong and where. Learning to read tracebacks is one of the most important skills in programming.

**How to read a traceback:**
1. **Start at the bottom** — the last line tells you the error type and message (e.g., `TypeError: ...`)
2. **Look at the arrow** (`--->`) — it points to the exact line that caused the error
3. **Read the file and line number** — this tells you where in your code the problem is

You can then copy the error message and search for it online (e.g., on Stack Overflow) to find solutions.

**Example — TypeError:**

````python
# This will raise a TypeError because sum() expects numbers, not strings
try:
    sum('test')
except TypeError as e:
    print("Error:", e)
    print("Tip: Search for 'TypeError: unsupported operand type(s) for +: 'int' and 'str'' online.")
````

**Example — NameError:**

A `NameError` happens when you try to use a variable or function that does not exist. This is often caused by a typo or forgetting to define something.

````python
# NameError: using a variable that was never defined
try:
    print(temperature_celsius)  # This variable does not exist
except NameError as e:
    print("Error:", e)
    print("Fix: Check your spelling or define the variable first.")
````

**Example — IndexError:**

An `IndexError` happens when you try to access an index that does not exist in a list (we learned about indexing in Notebook 8).

````python
# IndexError: accessing an index that is out of range
measurements = [10.5, 20.3, 30.1]  # Only indices 0, 1, 2 exist
try:
    print(measurements[5])  # Index 5 does not exist
except IndexError as e:
    print("Error:", e)
    print("Fix: Check the length of your list with len().")
    print("List length:", len(measurements))
````

### Summary: Common Python Error Types

| Error Type | What It Means | Common Cause |
|---|---|---|
| `TypeError` | Wrong data type used | Passing a string where a number is expected |
| `NameError` | Variable/function not found | Typo in variable name, or forgot to define it |
| `IndexError` | Index out of range | Accessing `list[5]` when list has only 3 items |
| `ValueError` | Right type, wrong value | `int("hello")` — string cannot become an integer |
| `AttributeError` | Object has no such method | Calling `.append()` on a tuple (tuples have no append) |
| `SyntaxError` | Invalid Python syntax | Missing colon `:`, unmatched parentheses |

## 5. Common Pitfalls

1. **Calling `help()` on the result instead of the function:**
   - Wrong: `help(print())` — this first calls `print()`, then passes `None` to `help()`
   - Correct: `help(print)` — pass the function itself without parentheses

2. **Only reading the first line of an error message:**
   - Always scroll to the **bottom** of the traceback — the actual error type and message are on the last line. The lines above show you the path the code took to get to the error.

3. **Ignoring the line number in the traceback:**
   - Python tells you exactly which line caused the error. Read the arrow (`--->`) and check that specific line first.

4. **Not searching for the exact error message:**
   - When searching online, copy the **full error message** (e.g., `TypeError: unsupported operand type(s) for +: 'int' and 'str'`). This gives you much better search results than just "Python error".

### Comparison: Help Tools at a Glance

| Tool | Where It Works | What It Shows | Best For |
|---|---|---|---|
| `help(obj)` | Everywhere | Full documentation | Detailed info in any environment |
| `obj?` | Jupyter / IPython only | Quick summary | Fast lookup while coding |
| `obj??` | Jupyter / IPython only | Source code (if available) | Seeing how something is implemented |
| `dir(obj)` | Everywhere | List of all methods/attributes | Discovering what is available |
| Tab completion | Jupyter / IPython only | Dropdown of methods | Quick browsing while typing |

## 6. Practice Tasks

**Task 1 (Easy):** Use `help()` to find out what the built-in `round()` function does. Then use `round()` to round the number `3.14159` to 2 decimal places. Print the result.

**Task 2 (Medium):** Use `dir()` on a string variable (e.g., `"hello"`) to find a method that converts the string to uppercase. Then call that method and print the result. (Hint: look for a method with "upper" in the name.)

**Task 3 (Hard):** The following code has a bug. Read the error message, identify the problem, and fix it:

```python
experiment_data = [23.1, 45.7, 12.3, 67.8]
average = sum(experiment_data) / len(experiment_data)
print("The average measurement is: " + average)
```

## 7. Task Solutions

````python
# Task 1 Solution
help(round)  # Read what round() does

# Round pi to 2 decimal places
pi_rounded = round(3.14159, 2)
print("Rounded value:", pi_rounded)  # Output: 3.14
````

````python
# Task 2 Solution
my_text = "hello"

# Use dir() to find string methods
methods = [m for m in dir(my_text) if not m.startswith('_')]
print("String methods:", methods)

# Found 'upper' in the list — let's use it
print("Uppercase:", my_text.upper())  # Output: HELLO
````

````python
# Task 3 Solution
# The bug: you cannot concatenate a string and a float with +
# Fix: convert the float to a string with str(), or use an f-string

experiment_data = [23.1, 45.7, 12.3, 67.8]
average = sum(experiment_data) / len(experiment_data)

# Option 1: using str()
print("The average measurement is: " + str(average))

# Option 2: using f-string (cleaner)
print(f"The average measurement is: {average:.2f}")
````
