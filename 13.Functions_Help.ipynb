## 1. Using `help()`
The `help()` function is the most standard way to access documentation in Python. It prints the docstring of the object you pass to it.

**Example:**

```python
alist = [1, 2, 3]

# Get help on the sum function
help(sum)

# Example usage based on help:
# sum(iterable, start=0)
print("Sum with start=10:", sum(alist, 10))
```

## 2. Using `?` and `??`
In Jupyter Notebooks and IPython, you can use `?` to get a quick summary of an object and `??` to see the source code (if available).


```python
# Get quick help
sum?

# Get source code (if available)
sum??
```

## 3. Inspecting Source Code
For more advanced users, the `inspect` module allows you to retrieve the source code of a function or class. This is useful for understanding how a library function is implemented.


```python
import numpy as np
import inspect

# Get the source lines of np.linspace
source_lines = inspect.getsourcelines(np.linspace)
print("Source code for np.linspace (first 5 lines):")
for line in source_lines[0][:5]:
    print(line.strip())
```

## 4. Troubleshooting Errors
When you encounter an error, the error message often provides a clue. You can copy the error message and search for it online (e.g., on Stack Overflow) to find solutions.

**Example Error:**

```python
# This will raise a TypeError because sum() expects numbers, not strings
try:
    sum('test')
except TypeError as e:
    print("Error:", e)
    print("Tip: Search for 'TypeError: unsupported operand type(s) for +: 'int' and 'str'' online.")
```
