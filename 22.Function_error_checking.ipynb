## Python Basics — Function Error Checking

In this notebook, we will learn how to make our functions more robust by adding **error checking**. We will explore how to validate inputs and handle exceptions gracefully.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `isinstance()` to check the type of function arguments.
- Raise custom exceptions using `raise Exception()`.
- Use `try-except` blocks to handle errors without crashing the program.
- Convert string inputs to numbers safely.


## 1. Basic Function (No Error Checking)
Let's start with a simple function that multiplies two numbers.

**Example:**

```python
def funfun(x, y):
    z = x * y
    return z

# Works fine with numbers
print("2 * 3 =", funfun(2, 3))

# But what if we pass a string?
# Python allows string multiplication, which might not be what we want.
print("2 * 'hi' =", funfun(2, 'hi'))
```

## 2. Adding Type Checking
We can use `isinstance()` to ensure that the inputs are of the expected type (e.g., `int` or `float`).


```python
def funfun2(x, y):
    # Check if x is an integer
    if not isinstance(x, int):
        raise Exception('Input x must be an integer!')
    z = x * y
    return z

# This will raise an error if x is not an integer
try:
    funfun2(2.5, 2)
except Exception as e:
    print("Error:", e)
```

## 3. Allowing Multiple Types
We can check for multiple types by passing a tuple to `isinstance()`.


```python
def funfun3(x, y):
    # Check if x is an integer or a float
    if not isinstance(x, (int, float)):
        raise Exception('Input x must be a number!')
    z = x * y
    return z

# Now it works with floats
print("2.5 * 2 =", funfun3(2.5, 2))
```

## 4. Converting Inputs
Sometimes we want to be flexible and accept strings that look like numbers. We can check if a string is numeric using `.isnumeric()` and convert it.


```python
def funfun4(x, y):
    # Check if x is numeric
    if not isinstance(x, (int, float)):
        # Check if x is a string that can be converted
        if isinstance(x, str) and x.isnumeric():
            x = float(x)
            print('Note: Converted x to float.')
        else:
            raise Exception('Input x must be a number!')

    z = x * y
    return z

# Test with a string number
print("Result:", funfun4('2', 3))

# Test with a non-numeric string (will raise error)
try:
    funfun4('asd2', 3)
except Exception as e:
    print("Error:", e)
```

## 5. Try-Except Blocks
The `try-except` block allows you to catch errors and handle them without stopping the program. This is useful for debugging or when you expect potential errors.


```python
def funfunfun(x, y):
    z = 0
    try:
        z = x**y  # Try to perform the operation
        print('Calculation successful.')
    except:
        print("Couldn't complete mission.")  # Handle the error
    return z

# Test with valid inputs
print("Result:", funfunfun(2, 4))

# Test with invalid inputs (string exponent)
print("Result:", funfunfun(2, 'asdf'))
```
