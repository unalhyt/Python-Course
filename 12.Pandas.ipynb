## 1. Importing Pandas
To use Pandas, you first need to import it. It is standard practice to alias it as `pd`. We will also import NumPy (`np`) for generating data.


```python
import pandas as pd
import numpy as np
```

## 2. Generating Data
We will generate some random data to simulate a dataset.

- `np.random.randn(n)` generates `n` random numbers from a standard normal distribution (mean=0, std=1).
- We can scale and shift these numbers to simulate real-world data (e.g., temperature).


```python
# Generate 100 random numbers for temperature (mean=20, std=5)
var1 = np.random.randn(100) * 5 + 20
print("Mean of var1:", np.mean(var1))

# Generate 100 boolean values (True/False) for ice cream consumption
# True if the random number is > 0, False otherwise
var2 = np.random.randn(100) > 0
```

## 3. Creating a DataFrame
A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types. We can create a DataFrame from a dictionary where keys are column names and values are lists or arrays of data.


```python
labels = ['Temp. (C)', 'Ice cream eaten']

# Create a dictionary
D = {labels[0]: var1, labels[1]: var2}

# Create a DataFrame
df = pd.DataFrame(data=D)
print("DataFrame created successfully.")
```

## 4. Inspecting Data
Pandas provides several methods to inspect the data in a DataFrame.

- `.head(n)`: Returns the first `n` rows.
- `.count()`: Returns the number of non-null values in each column.
- `.mean()`: Returns the mean of numeric columns.


```python
# Display the first 10 rows
print(df.head(10))

# Count non-null values
print("\nCounts:\n", df.count())

# Calculate mean values
print("\nMeans:\n", df.mean())
```

## 6. Why Pandas? (NumPy vs Pandas)

You already saw **NumPy** in Notebook 11. A NumPy array is great for numbers but has no column names or row labels. A **Pandas DataFrame** is like a NumPy array **wrapped in a spreadsheet** — you get labeled rows, labeled columns, mixed data types, and lots of convenience methods.

**Analogy:** NumPy is a calculator. Pandas is a calculator sitting inside Excel.

| Feature | NumPy array | Pandas DataFrame |
|---|---|---|
| Row/column labels | No | Yes |
| Mixed types per column | No (all same) | Yes |
| Missing values (NaN) | Awkward | Native support |
| Summary stats | `np.mean()` etc. | `.describe()`, `.value_counts()` |
| Load CSV / Excel | Hard | One-line |
| Best for | Pure numeric math | Data analysis & tables |

## 7. Series — The Building Block of a DataFrame

Before a DataFrame (a full table) there is a **Series** — a single column with labels. Think of a Series as a NumPy array plus an **index** (row labels). A DataFrame is basically a dictionary of Series that share the same index.

```python
# Series from a list (default integer index)
s1 = pd.Series([10, 20, 30, 40])
print(s1)

# Series from a dict: keys become the index labels
s2 = pd.Series({"a": 1.5, "b": 2.8, "c": 3.1})
print("\nWith labels:\n", s2)
print("\nValue at label 'b':", s2["b"])
```

## 8. Selecting Columns and Rows

There are three main ways to pick data out of a DataFrame:

- `df["col"]` → one column (a Series).
- `df[["col1", "col2"]]` → several columns (a smaller DataFrame). **Note the double brackets.**
- `df.loc[row_label, col_label]` → by **label**.
- `df.iloc[row_pos, col_pos]` → by **integer position** (see Notebooks 8 & 9 for the same idea on lists/arrays).

```python
# Select one column -> returns a Series
temps = df["Temp. (C)"]
print("Type:", type(temps).__name__)
print(temps.head(3))

# Select multiple columns -> returns a DataFrame
sub = df[["Temp. (C)", "Ice cream eaten"]]
print("\nSub-DataFrame:\n", sub.head(3))

# Row access: .loc uses labels, .iloc uses integer positions
print("\nFirst row (iloc):\n", df.iloc[0])
print("\nRow with label 5 (loc):\n", df.loc[5])
```

## 9. Filtering Rows (Boolean Masks)

A comparison like `df["Temp. (C)"] > 25` produces a Series of `True`/`False`. Passing that back into `df[...]` keeps only the rows where the value is `True`. This is called **boolean masking** and is how filtering works in Pandas.

Combine conditions with `&` (AND) and `|` (OR). **Always wrap each condition in parentheses.**

```python
# Boolean mask: keep only hot days
hot_days = df[df["Temp. (C)"] > 25]
print("Hot days count:", len(hot_days))
print(hot_days.head())

# Combine conditions with & (AND) and | (OR) — brackets around each condition!
hot_and_ice = df[(df["Temp. (C)"] > 25) & (df["Ice cream eaten"])]
print("\nHot days where ice cream was eaten:", len(hot_and_ice))
```

## 10. Adding and Modifying Columns

Assigning to a new column name creates it. You can compute a new column from existing ones using vectorized math (the same idea as NumPy in Notebook 11).

```python
# Add a brand new column: temperature in Fahrenheit
df["Temp. (F)"] = df["Temp. (C)"] * 9/5 + 32     # vectorized, like NumPy
print(df.head())

# Add a column from a condition (hot day flag)
df["hot_day"] = df["Temp. (C)"] > 25              # True if over 25 C
print("\nAfter adding hot_day:\n", df.head())
```

## 11. Summary Statistics

`df.describe()` gives a one-shot report of every numeric column: count, mean, std, min, 25%, 50%, 75%, max. You can also call stats on a single column.

For categorical or boolean columns, `.value_counts()` tells you how often each value appears.

```python
# describe() gives count, mean, std, min, quartiles, max
print(df.describe())

# Per-column stats
print("\nMax temp:", df["Temp. (C)"].max())
print("Std temp:", df["Temp. (C)"].std())

# value_counts for categorical (boolean) columns
print("\nIce cream counts:\n", df["Ice cream eaten"].value_counts())
```

## 12. Sorting Rows

Use `df.sort_values(by="col")` to order rows by a column. `ascending=False` flips it to descending. You can pass a list of column names to break ties.

```python
# Sort the DataFrame by temperature (highest first)
sorted_df = df.sort_values(by="Temp. (C)", ascending=False)
print(sorted_df.head())

# Sort by multiple columns: ice cream first, then temperature
multi_sorted = df.sort_values(by=["Ice cream eaten", "Temp. (C)"],
                              ascending=[False, False])
print("\nMulti-sort head:\n", multi_sorted.head())
```

## 13. Reading & Writing CSV Files

Most real datasets live in CSV files (comma-separated values). Pandas can **read** and **write** them in one line.

- `pd.read_csv("data.csv")` → loads a CSV into a DataFrame.
- `df.to_csv("out.csv", index=False)` → writes a DataFrame to a CSV (no row index column).

```python
# Write a DataFrame to CSV and read it back
small = pd.DataFrame({"x": [1, 2, 3], "y": [10, 20, 30]})

small.to_csv("demo.csv", index=False)    # save to a file, no row index column
loaded = pd.read_csv("demo.csv")         # read the file back
print(loaded)
```

## 14. Engineering Example 1 — Beam Deflection Table (Civil / Mechanical Engineering)

A **civil engineer** tests different loads on a steel beam and records the center deflection. We store the result in a DataFrame and filter the rows that exceed a safety threshold.

```python
# Civil / mechanical: simply-supported beam, center deflection vs load
# Formula (point load at center): delta = P * L^3 / (48 * E * I)
L  = 2.0            # beam length in meters
E  = 200e9          # Young's modulus (steel), Pa
I  = 8e-6           # second moment of area, m^4

loads_kn = np.array([1, 2, 5, 10, 20, 50, 100])               # load in kN
deflect_mm = (loads_kn*1000) * L**3 / (48 * E * I) * 1000     # in mm

beam = pd.DataFrame({"load_kN": loads_kn, "deflection_mm": deflect_mm.round(3)})
print(beam)

# Which loads cause deflection greater than 1 mm (a safety threshold)?
print("\nUnsafe loads:\n", beam[beam["deflection_mm"] > 1.0])
```

## 15. Engineering Example 2 — Reactor Sensor Log (Chemical / Process Engineering)

A **process engineer** logs temperature and pressure from a reactor every minute. Pandas makes it easy to store, summarize, and flag out-of-range samples.

```python
# Chemical / process engineering: reactor sensor log
np.random.seed(0)                                             # reproducible demo
log = pd.DataFrame({
    "minute":   np.arange(10),
    "temp_c":   70 + np.random.randn(10) * 2,                 # around 70 C
    "pressure": 5 + np.random.randn(10) * 0.3,                # around 5 bar
})
print(log.round(2))

# Flag outliers: temperature more than 2 C away from the mean
mean_t = log["temp_c"].mean()
log["temp_alarm"] = np.abs(log["temp_c"] - mean_t) > 2        # True = out of range
print("\nAlarm rows:\n", log[log["temp_alarm"]].round(2))
```

## 16. Common Pitfalls

1. **Spaces in column names.** `df.Temp. (C)` won't work. Use brackets: `df["Temp. (C)"]`.
2. **Chained assignment warning.** `df[df.x > 0]["y"] = 5` may show `SettingWithCopyWarning` and silently fail. Use `.loc`: `df.loc[df.x > 0, "y"] = 5`.
3. **Filter creates a copy.** `big = df[df.x > 0]` is a new DataFrame. Modifying `big` does not change `df`.
4. **`.mean()` on non-numeric columns.** Taking the mean of a string column raises an error. Select numeric columns first or use `df.mean(numeric_only=True)`.
5. **`.loc` vs `.iloc`.** `.loc[0]` uses the **label** `0`. `.iloc[0]` uses the **first row** regardless of label. If your index was reset, they often look the same but aren't.
6. **Forgetting to assign the result.** `df.sort_values("x")` does **not** modify `df`. Save it: `df = df.sort_values("x")` or pass `inplace=True`.

## 19. Real Dataset Walkthrough — Titanic Passenger List

Until now we built DataFrames by hand. Real data analysis starts by **loading a file**. Let's use the **Titanic passenger CSV** from the official Pandas tutorial — the same file used across the pandas.pydata.org "Getting started" pages.

**Source:** `https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv`

It has 891 passengers with columns for name, age, sex, ticket class (`Pclass`), fare, whether they survived (`Survived`), and more. We will use **this single file** to practice every concept from the sections above: loading, inspecting, selecting, filtering, adding columns, summary stats, sorting, and grouping.

### 19.1 Load the File

Pandas can read a CSV directly from a URL — no manual download required.

```python
# Step 1 — Load the CSV
# Option A (used here): read straight from a URL on the internet
url = "https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv"
titanic = pd.read_csv(url)      # one line: fetch and parse CSV into a DataFrame

# Option B: load the same file from your local computer
# The file "titanic.csv" is saved in the same folder as this notebook.
# If the notebook is in a different folder, give the full path, e.g.:
#   titanic = pd.read_csv(r"C:\Users\unalh\PycharmProjects\Python-Course\titanic.csv")
# titanic = pd.read_csv("titanic.csv")        # relative path (same folder)

# Column headings (what each column means):
#   PassengerId -> unique row number for each passenger (1..891)
#   Survived    -> 0 = died, 1 = survived
#   Pclass      -> ticket class: 1 = 1st (upper), 2 = 2nd (middle), 3 = 3rd (lower)
#   Name        -> passenger's full name
#   Sex         -> "male" or "female"
#   Age         -> age in years (may be missing / NaN)
#   SibSp       -> number of siblings or spouses aboard
#   Parch       -> number of parents or children aboard
#   Ticket      -> ticket number (text code)
#   Fare        -> ticket price paid, in British pounds
#   Cabin       -> cabin number (often missing / NaN)
#   Embarked    -> port of boarding: C = Cherbourg, Q = Queenstown, S = Southampton

print(titanic.head())           # peek at the first 5 rows
```

### 19.2 Inspect the Data

Whenever you load a new dataset, your **first four checks** should always be: shape, column names, data types, and missing values.

```python
# Step 2 — Inspect the data
print("Shape (rows, cols):", titanic.shape)      # how big is the table
print("\nColumn names:", list(titanic.columns))
print("\nData types per column:\n", titanic.dtypes)
print("\nMissing values per column:\n", titanic.isna().sum())
```

### 19.3 Selecting Columns and Filtering Rows

Now let's use the concepts from §8 and §9 on real data.

```python
# Step 3 — Select columns and filter rows

# Pick just a few columns we care about
subset = titanic[["Name", "Sex", "Age", "Pclass", "Survived"]]
print(subset.head())

# Filter: all female passengers who survived
survived_women = titanic[(titanic["Sex"] == "female") & (titanic["Survived"] == 1)]
print("\nFemale survivors:", len(survived_women))

# Filter: children younger than 12
children = titanic[titanic["Age"] < 12]
print("\nChildren (age < 12):", len(children))
print(children[["Name", "Age", "Survived"]].head())
```

### 19.4 Adding a Computed Column and Summary Stats

Let's create a new column that combines two existing ones, then use `.describe()` and `.value_counts()` to summarize.

```python
# Step 4 — Add a computed column + summary stats
# SibSp = siblings/spouses aboard; Parch = parents/children aboard
titanic["FamilySize"] = titanic["SibSp"] + titanic["Parch"] + 1   # +1 for self
print(titanic[["Name", "SibSp", "Parch", "FamilySize"]].head())

# Numeric summary of the whole table
print("\nDescribe:\n", titanic.describe().round(2))

# How many passengers per class? value_counts is perfect for categorical columns
print("\nPassengers per class:\n", titanic["Pclass"].value_counts())
```

### 19.5 Sorting and Grouping

Who paid the most for their ticket? Who had the best chances of survival? `sort_values` and `groupby` answer these.

```python
# Step 5 — Sorting and grouping on the Titanic data
# Top 5 most expensive tickets
top_fares = titanic.sort_values("Fare", ascending=False).head(5)
print("Top 5 fares:\n", top_fares[["Name", "Pclass", "Fare", "Survived"]])

# Survival rate by passenger class (Pclass): group rows and take the mean of Survived
survival_by_class = titanic.groupby("Pclass")["Survived"].mean()
print("\nSurvival rate by class:\n", survival_by_class.round(3))

# Survival rate by sex
survival_by_sex = titanic.groupby("Sex")["Survived"].mean()
print("\nSurvival rate by sex:\n", survival_by_sex.round(3))
```

### 19.6 Extra Statistics — Pulling It All Together

Let's dig deeper into the Titanic data using **only the skills learned so far**: NumPy math, lists/dicts, booleans (Notebook 6), filtering, `.mean` / `.min` / `.max`, `groupby`, `value_counts`, and string-formatted printing.

**What each statistics command does:**

- `.dropna()` — returns a copy of the Series with missing values (NaN) removed. Needed because `Age` has 177 blanks in this file.
- `.count()` — how many **non-missing** values the column has.
- `.mean()` — arithmetic average: sum of all values divided by count.
- `.median()` — the middle value when the data is sorted (less sensitive to outliers than mean).
- `.std()` — standard deviation: how spread out the values are from the mean. Small std = values cluster near the mean; large std = values are scattered.
- `.min()` / `.max()` — smallest and largest value.
- `(ages < 12).sum()` — the comparison produces a Series of `True`/`False`. In Python, `True` counts as `1` and `False` as `0` (Notebook 6), so summing them gives the number of `True`s.
- `.groupby("col")` — splits the table into groups that share the same value in `col`. Any stat you call afterwards is computed **per group**.
- `.agg(["mean", "min", "max"])` — apply several stats at once and return them as a small table.
- `.value_counts()` — counts how many times each unique value appears in a column (great for text / categorical data).
- `df["Survived"].mean() * 100` — since `Survived` is 0/1, the mean is the fraction that survived; multiplying by 100 converts to a percentage.

```python
# Step 6a — Basic numeric statistics on the Age column
# Uses only skills learned so far: NumPy math, indexing, booleans, .mean/.min/.max
ages = titanic["Age"].dropna()                   # drop missing ages

print("Passenger count (with Age):", ages.count())
print("Mean age:   ", round(ages.mean(), 2))
print("Median age: ", round(ages.median(), 2))
print("Std dev:    ", round(ages.std(), 2))
print("Min age:    ", ages.min())
print("Max age:    ", ages.max())

# How many passengers were children (< 12) or seniors (>= 60)?
# Boolean mask + sum counts True values (from Notebook 6).
n_children = (ages < 12).sum()
n_seniors  = (ages >= 60).sum()
print(f"\nChildren (<12): {n_children}")
print(f"Seniors (>=60): {n_seniors}")
```
