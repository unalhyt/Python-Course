## Python Basics — Plotting Dots & Lines

In this notebook, we will explore **Matplotlib**, the most popular plotting library in Python. We will learn how to create simple plots, including scatter plots (dots) and line plots.

## Learning Goals
By the end of this notebook, students will be able to:
- Import `matplotlib.pyplot` as `plt`.
- Create scatter plots using `plt.plot()` with markers (e.g., `'o'`, `'s'`).
- Customize plots with colors (e.g., `'r'`, `'b'`) and labels.
- Add legends using `plt.legend()`.
- Plot lines by specifying start and end coordinates.
- Plot mathematical functions (e.g., sine wave).


## 1. Importing Matplotlib
To use Matplotlib, you first need to import the `pyplot` module. It is standard practice to alias it as `plt`.


```python
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Plotting Dots (Scatter Plots)
You can plot individual points by specifying their x and y coordinates and a marker style.

- `'o'`: Circle
- `'s'`: Square
- `'p'`: Pentagon
- `'r'`: Red color
- `'b'`: Blue color


```python
# Plot a red circle at (3, 2)
plt.plot(3, 2, 'ro', label='Red Circle')

# Plot a blue pentagon at (3, 4)
plt.plot(3, 4, 'bp', label='Blue Pentagon')

# Add a legend to identify the points
plt.legend()

# Display the plot
plt.show()
```

## 3. Plotting Lines
To plot a line, you provide lists of x-coordinates and y-coordinates. Matplotlib connects the points in order.

Syntax: `plt.plot([x1, x2], [y1, y2], 'color')`


```python
# Plot a red line from (0, 0) to (1, 2)
plt.plot([0, 1], [0, 2], 'r', label='Red Line')

# Plot a green line from (0, 1) to (0, 2)
plt.plot([0, 0], [1, 2], 'g', label='Green Line')

plt.legend()
plt.show()
```

## 4. Plotting Functions
You can plot mathematical functions by generating arrays of x and y values using NumPy.


```python
# Generate x values from 0 to 3*pi
x = np.linspace(0, 3 * np.pi, 100)

# Calculate y values (sine wave)
y = np.sin(x)

# Plot the sine wave with black dots ('k.')
plt.plot(x, y, 'k.')
plt.title("Sine Wave")
plt.show()
```

Draw your favorite letter using lines!

Example: Drawing the letter 'X'.


```python
# Draw the first diagonal of 'X' (from 0,0 to 1,1)
plt.plot([0, 1], [0, 1], 'm', linewidth=20)

# Draw the second diagonal of 'X' (from 0,1 to 1,0)
plt.plot([0, 1], [1, 0], 'm', linewidth=20)

# Set the aspect ratio to be square so the letter isn't distorted
plt.axis('square')

plt.show()
```
