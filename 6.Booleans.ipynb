## Python Basics — Booleans

In this notebook, we will explore **Booleans**, a fundamental data type in Python that represents truth values: `True` or `False`. Booleans are essential for making decisions in your code.

## Learning Goals
By the end of this notebook, students will be able to:
- Understand the concept of Boolean values (`True` and `False`).
- Use comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) to generate Booleans.
- Combine multiple conditions using logical operators (`and`, `or`, `not`).
- Convert other data types to Booleans using `bool()`.
- Apply Boolean logic to solve simple problems.


## 1. Boolean Values
A Boolean value can only be `True` or `False`. Note that the first letter must be capitalized.

**Example:**

```python
boolTrue = True
strTrue = 'True'  # This is a string, not a boolean

print("Boolean True:", boolTrue)
print("String 'True':", strTrue)

# Check types
print("Type of boolTrue:", type(boolTrue))
print("Type of strTrue:", type(strTrue))
```

## 2. Comparison Operators
Comparison operators allow you to compare two values. The result of a comparison is always a Boolean.

- `==`: Equal to
- `!=`: Not equal to
- `>`: Greater than
- `<`: Less than
- `>=`: Greater than or equal to
- `<=`: Less than or equal to


```python
# Is 4 equal to 4?
print("4 == 4:", 4 == 4)

# Is 4 greater than or equal to 5?
print("4 >= 5:", 4 >= 5)

# Is 4 NOT equal to 5?
print("4 != 5:", 4 != 5)

# Using variables
a = 8
b = 5
outcome = a == b * 2  # Is 8 equal to 10?
print("Outcome:", outcome)
```

## 3. Logical Operators
You can combine multiple conditions using `and`, `or`, and `not`.

- `and`: Returns `True` if **both** conditions are true.
- `or`: Returns `True` if **at least one** condition is true.
- `not`: Reverses the boolean value.


```python
# Using 'and'
print("14 > 5 and 7 < 10 and 3 == 3:", 14 > 5 and 7 < 10 and 3 == 3)

# Using 'or'
print("14 > 5 or 7 > 10:", 14 > 5 or 7 > 10)
```

## 4. Converting to Booleans
You can convert other data types to Booleans using the `bool()` function.

- Numbers: `0` converts to `False`, any other number converts to `True`.
- Strings: An empty string `""` converts to `False`, any other string converts to `True`.


```python
print("bool(0):", bool(0))
print("bool(1):", bool(1))
print("bool(123.45):", bool(123.45))

print("bool('asd'):", bool('asd'))
print("bool(' '):", bool(' '))  # Space is not empty
print("bool(''):", bool(''))    # Empty string
```
