## Python Basics — Functions: Input & Output

In this notebook, we will explore **functions**, which are reusable blocks of code that perform a specific task. We will focus on how to pass data (input) to functions and how to get results (output) from them.

Think of a function like a **machine in a factory**:
- You feed **raw materials** into the machine (these are the **arguments / inputs**)
- The machine does some work (the **function body**)
- The machine produces a **finished product** (the **return value / output**)

For example, `sum([10, 20, 30])` feeds the list into the "sum machine" and gets `60` out.

## 1. Using Built-in Functions
Python comes with many built-in functions that are ready to use. You do not need to install or import anything — they are always available.

Think of built-in functions as **tools in a toolbox** that Python gives you for free. Each tool has a specific job:

| Function | What It Does | Example |
|----------|-------------|--------|
| `sum()` | Adds up all elements | `sum([1, 2, 3])` → `6` |
| `len()` | Counts elements | `len([1, 2, 3])` → `3` |
| `max()` | Finds the largest value | `max([1, 2, 3])` → `3` |
| `min()` | Finds the smallest value | `min([1, 2, 3])` → `1` |
| `abs()` | Removes the negative sign | `abs(-5)` → `5` |
| `round()` | Rounds a number | `round(3.14159, 2)` → `3.14` |
| `sorted()` | Returns a sorted list | `sorted([3, 1, 2])` → `[1, 2, 3]` |

**Example:**

````python
# Calculate the sum of a list directly
print("Sum of [10, 20, 30]:", sum([10, 20, 30]))

# Using a variable as input
lst = [10, 20, 30]
total = sum(lst)
print("Sum of lst:", total)

# Note: Be careful with parentheses! Functions are called using ().
````

````python
# Experiment: temperature readings from a sensor (in Celsius)
temperatures = [22.5, 19.8, 25.1, 18.3, 23.7]

# Find the highest temperature recorded
print("Max temperature:", max(temperatures))  # 25.1

# Find the lowest temperature recorded
print("Min temperature:", min(temperatures))  # 18.3

# Sort temperatures from lowest to highest
print("Sorted:", sorted(temperatures))
````

````python
# abs() removes the negative sign — useful for error calculations
measured_value = 9.78       # measured gravity (m/s^2)
expected_value = 9.81       # expected gravity (m/s^2)

# Calculate absolute error
error = measured_value - expected_value  # negative result
print("Error:", error)                   # -0.03
print("Absolute error:", abs(error))     # 0.03
````

````python
# round() controls decimal places
pi = 3.14159265
print("Pi rounded to 2 decimals:", round(pi, 2))  # 3.14
print("Pi rounded to 4 decimals:", round(pi, 4))  # 3.1416
print("Pi rounded to 0 decimals:", round(pi))      # 3
````

## 2. Function Arguments
Functions can take one or more **arguments** (inputs). An argument is the data you pass inside the parentheses when you call a function.

- Some functions take **one argument**: `len(lst)`, `sum(lst)`, `abs(-5)`
- Some functions take **multiple arguments**: `round(3.14159, 2)`, `pow(2, 3)`
- The **order** of arguments matters — `round(3.14, 2)` is not the same as `round(2, 3.14)`

Think of arguments like **ingredients in a recipe**. The function is the recipe, and the arguments are what you put in. If you put in the wrong ingredients (or the wrong number), the recipe will not work!

````python
# Calculate the length of the list
list_length = len(lst)
print("Length of lst:", list_length)

# Some functions can take multiple arguments, which we will see later.
````

````python
# Functions with multiple arguments

# pow(base, exponent) — raises base to the power of exponent
print("2 to the power of 10:", pow(2, 10))  # 1024

# round(number, decimal_places)
speed = 343.456789  # speed of sound in m/s
print("Rounded to 1 decimal:", round(speed, 1))  # 343.5
print("Rounded to 0 decimals:", round(speed, 0))  # 343.0
````

````python
# max() and min() can take a list OR multiple separate arguments
print("Max of a list:", max([5, 12, 3]))   # 12
print("Max of values:", max(5, 12, 3))     # 12

print("Min of a list:", min([5, 12, 3]))   # 3
print("Min of values:", min(5, 12, 3))     # 3
````

## 3. Combining Functions
You can use the output of one function as the input to another, or combine their results in an expression. This is called **nesting** functions.

When Python sees nested functions, it evaluates from the **inside out** — just like math:
```
round(sum(lst) / len(lst), 2)
  → round(60 / 3, 2)
  → round(20.0, 2)
  → 20.0
```

````python
# Calculate the average of the list
# Average = Sum / Count
average = sum(lst) / len(lst)

print("Average of lst:", average)
````

````python
# More examples of combining (nesting) functions

# Experiment: pressure readings in kPa
pressures = [101.325, 99.8, 102.1, 100.5, 98.7]

# Nested: round the average to 2 decimal places
avg_pressure = round(sum(pressures) / len(pressures), 2)
print("Average pressure:", avg_pressure, "kPa")

# Nested: find the range (max - min)
pressure_range = round(max(pressures) - min(pressures), 2)
print("Pressure range:", pressure_range, "kPa")
````

````python
# You can also sort and then pick values using slicing
# (We learned about slicing in Notebook 9)
sorted_pressures = sorted(pressures)
print("All sorted:", sorted_pressures)
print("Lowest two readings:", sorted_pressures[:2])
print("Highest two readings:", sorted_pressures[-2:])
````

## 4. Return Values
Every function **returns** something. The return value is the "output" of the function — the finished product from our factory machine analogy.

- `sum([1, 2, 3])` **returns** `6` — you can store it in a variable
- `print("hello")` **returns** `None` — it displays text on screen but gives back nothing useful

This is an important distinction: **`print()` is NOT the same as returning a value!**

You can capture a return value by assigning it to a variable:

````python
# sum() returns a number — we can store and use it
result = sum([100, 200, 300])
print("The result is:", result)       # 600
print("Double the result:", result * 2)  # 1200
````

````python
# print() returns None — it only displays text
x = print("Hello")  # This prints "Hello" on screen
print("x is:", x)   # But x is None, not "Hello"!
print("Type of x:", type(x))  # <class 'NoneType'>
````

````python
# sorted() returns a NEW list — it does not change the original
grades = [85, 92, 78, 95, 88]
sorted_grades = sorted(grades)  # store the return value

print("Original:", grades)        # unchanged
print("Sorted copy:", sorted_grades)  # new sorted list
````

## 5. Type Conversion Functions
Python has built-in functions that convert data from one type to another. These are very useful when you need to change a number to text, or text to a number.

We first learned about types in **Notebook 1 (Variables & Types)**. Here we see them as functions with input and output:

| Function | Converts To | Example |
|----------|-----------|--------|
| `int()` | Integer | `int(3.7)` → `3` |
| `float()` | Decimal number | `float(5)` → `5.0` |
| `str()` | Text string | `str(42)` → `"42"` |
| `type()` | Shows the type | `type(3.14)` → `<class 'float'>` |

````python
# Convert float to int (truncates, does NOT round)
altitude = 10432.7  # meters
print("As integer:", int(altitude))    # 10432 (decimal part removed)
print("Rounded:", round(altitude))     # 10433 (properly rounded)
````

````python
# Convert int to float
distance = 500  # meters
print("As float:", float(distance))  # 500.0

# Convert number to string (useful for building messages)
speed = 340  # m/s
message = "The speed of sound is " + str(speed) + " m/s"
print(message)
````

````python
# Convert string to number (useful when reading data)
reading = "25.4"  # this is text, not a number
print("Type before:", type(reading))  # str

value = float(reading)  # now it is a number
print("Type after:", type(value))     # float
print("Value + 10:", value + 10)      # 35.4
````

## 6. Common Pitfalls

Here are common mistakes students make when using functions:

**1. Forgetting the parentheses**

Without `()`, you get the function object itself, not its result.

````python
numbers = [1, 2, 3]

# WRONG: forgot parentheses — this shows the function object
print("Without ():", len)  # <built-in function len>

# CORRECT: call the function with ()
print("With ():", len(numbers))  # 3
````

**2. Passing the wrong type of argument**

Functions expect specific types. `sum()` needs a list of numbers, not a single number.

````python
# WRONG: sum() expects a list, not separate numbers
# sum(10, 20, 30)  # This would cause a TypeError!

# CORRECT: pass a list to sum()
print("Correct sum:", sum([10, 20, 30]))  # 60

# Note: max() and min() CAN take separate numbers
print("Max works both ways:", max(10, 20, 30))  # 30
````

**3. Confusing `print()` with a return value**

`print()` displays text but returns `None`. Do not use it to "save" a result.

````python
# WRONG: trying to save print output
result = print(5 + 3)   # prints 8, but result is None
print("result is:", result)  # None

# CORRECT: save the calculation, then print it
result = 5 + 3           # result is 8
print("result is:", result)  # 8
````

**4. Confusing `int()` truncation with rounding**

`int()` always cuts off the decimal part. Use `round()` if you want proper rounding.

````python
value = 3.9

# int() truncates (cuts off decimal) — always goes toward zero
print("int(3.9):", int(value))    # 3  (not 4!)

# round() rounds to nearest integer
print("round(3.9):", round(value))  # 4
````

**5. Converting non-numeric strings to numbers**

`int()` and `float()` only work on strings that look like numbers.

````python
# This works — the string contains a valid number
print(float("3.14"))  # 3.14

# This would cause an error:
# float("hello")  # ValueError: could not convert string to float
# int("12.5")     # ValueError: invalid literal (use float() first)
````

## 8. Task Solutions

````python
# Solution 1: Average distance
distances = [120.5, 98.3, 145.7, 110.2, 132.8]

average_distance = sum(distances) / len(distances)
print("Average distance:", average_distance, "m")
````

````python
# Solution 2: Highest, lowest, and range
scores = [72, 85, 91, 68, 77, 95, 83]

print("Highest score:", max(scores))
print("Lowest score:", min(scores))
print("Range:", max(scores) - min(scores))
````

````python
# Solution 3: Nested functions with voltage data
voltages = [4.827, 5.102, 4.956, 5.211, 4.889]

# Average rounded to 2 decimals (one nested line)
avg_voltage = round(sum(voltages) / len(voltages), 2)
print("Average voltage:", avg_voltage, "V")

# Absolute difference between max and min
voltage_range = round(abs(max(voltages) - min(voltages)), 2)
print("Voltage range:", voltage_range, "V")
````

````python
# Solution 4: Convert strings to floats and sum them
reading_1 = "23.5"
reading_2 = "18.9"
reading_3 = "31.2"

# Convert each string to a float
measurements = [float(reading_1), float(reading_2), float(reading_3)]
print("Measurements:", measurements)
print("Total:", sum(measurements))
````

````python
# Solution 5: Wind tunnel analysis
speeds = [52.3, 48.7, 55.1, 49.8, 53.6, 51.2, 47.9, 54.4]
expected_speed = 51.5  # m/s

# 1. Number of measurements
print("Number of measurements:", len(speeds))

# 2. Average speed rounded to 1 decimal
avg_speed = round(sum(speeds) / len(speeds), 1)
print("Average speed:", avg_speed, "m/s")

# 3. Sorted list
print("Sorted speeds:", sorted(speeds))

# 4. Absolute error from expected speed
abs_error = round(abs(avg_speed - expected_speed), 2)
print("Absolute error:", abs_error, "m/s")
````
