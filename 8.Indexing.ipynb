## Python Basics — Indexing

In this notebook, we will explore **indexing**, a fundamental concept in Python for accessing specific elements within sequences like lists, strings, and tuples.

## Learning Goals
By the end of this notebook, students will be able to:
- Understand that Python uses **zero-based indexing**.
- Access individual elements in a list or string using square brackets `[]`.
- Use negative indexing to access elements from the end of a sequence.
- Access elements within nested lists (lists inside lists).
- Recognize that lists cannot be indexed with another list (unlike NumPy arrays).


## 1. Zero-Based Indexing
Python starts counting from **0**, not 1. The first element is at index `0`, the second at index `1`, and so on.

**Think of it like this:** Imagine a row of mailboxes numbered starting from 0. The first mailbox is mailbox 0, the second is mailbox 1, and so on. This is how Python labels positions inside a list, string, or tuple.

Here is a visual diagram for a list with 5 elements:

```
List:    [ 10,  11,  12,  13,  14 ]
Index:     0    1    2    3    4
```

**Example:**

````python
# Create a range of numbers and convert to a list
arange = range(10, 15)
alist = list(arange)
print("List:", alist)

# Access the second element (index 1)
print("Element at index 1:", alist[1])
````

````python
# Another example: list of planet names
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter"]

# Access the first planet (index 0)
print("First planet (index 0):", planets[0])

# Access the third planet (index 2)
print("Third planet (index 2):", planets[2])

# Access the fifth planet (index 4)
print("Fifth planet (index 4):", planets[4])
````

## 2. Negative Indexing
Python allows you to access elements from the **end** of a list using negative numbers. `-1` refers to the last element, `-2` to the second-to-last, etc.

**Think of it like this:** Imagine you are standing at the end of a line of people. The person right in front of you is at position `-1`, the one before them is `-2`, and so on.

Here is a visual diagram showing both positive and negative indices:

```
List:           [ 54,  63,  65,   9,   0, -644, 524 ]
Positive index:    0    1    2    3    4    5     6
Negative index:   -7   -6   -5   -4   -3   -2    -1
```

Notice that `list[-1]` always gives you the **last** element, no matter how long the list is.

````python
# Negative indexing is the same as using len() minus a number
alist = [54, 63, 65, 9, 0, -644, 524]

# These two are equivalent:
print("Using -1:", alist[-1])
print("Using len()-1:", alist[len(alist) - 1])

# These two are also equivalent:
print("Using -3:", alist[-3])
print("Using len()-3:", alist[len(alist) - 3])
````

````python
alist = [54, 63, 65, 9, 0, -644, 524]
print("List:", alist)

# Access the last element
print("Last element (-1):", alist[-1])

# Access the third-to-last element
print("Third-to-last element (-3):", alist[-3])
````

## 3. Using Variables as Indices
You can use a variable to specify the index. This is very useful when you want to access different elements depending on some calculation or condition.

You can also use expressions like `len()` to compute an index dynamically.

> **Note:** You cannot use a list as an index for a standard Python list. To select multiple elements at once, you need **NumPy arrays** (see Notebook 11).

````python
idx = 3
print("Element at index", idx, ":", alist[idx])

# Note: You cannot use a list as an index for a standard Python list.
# idx = [3, 4]
# alist[idx]  # This would raise a TypeError
````

````python
# Using len() to find the middle element of a list
temperatures = [15.2, 18.7, 22.1, 25.4, 19.8]

# Calculate the middle index
middle_idx = len(temperatures) // 2  # // is integer division
print("Middle index:", middle_idx)
print("Middle temperature:", temperatures[middle_idx])
````

## 4. Nested Indexing
When you have lists inside lists (or other nested structures), you can chain indices to access deeper elements.

**Think of it like this:** Imagine a filing cabinet (outer list). Each drawer (inner list) contains folders. To find a specific folder, you first pick the drawer number, then the folder number inside that drawer.

**How chained indexing works step by step:**
- `alist[2]` → gets the inner list `[9, 4, 6]`
- `alist[2][2]` → gets element at index 2 from that inner list → `6`

**Example:**

````python
alist = [4, 3.2, [9, 4, 6], 32]
print("Nested list:", alist)

# Access the inner list at index 2
inner_list = alist[2]
print("Inner list:", inner_list)

# Access the third element (index 2) of the inner list
element = inner_list[2]
print("Element from inner list:", element)

# Shortcut: Chain the indices
print("Direct access:", alist[2][2])
````

````python
# Deeper nesting: list of lists of lists
# Experiment data: 2 experiments, each with 2 trials, each with 3 measurements
experiments = [
    [[1.2, 1.3, 1.1], [2.4, 2.5, 2.3]],  # Experiment 0
    [[3.6, 3.7, 3.5], [4.8, 4.9, 4.7]]   # Experiment 1
]

# Get experiment 1, trial 0, measurement 2
value = experiments[1][0][2]
print("Experiment 1, Trial 0, Measurement 2:", value)

# Step by step:
print("Step 1 - experiments[1]:", experiments[1])
print("Step 2 - experiments[1][0]:", experiments[1][0])
print("Step 3 - experiments[1][0][2]:", experiments[1][0][2])
````

## 5. Indexing Strings
Strings are sequences of characters, so you can index them just like lists. Each character has a position number.

However, unlike lists, **strings are immutable** — you cannot change a character by assigning to an index. This is similar to tuples (see Notebook 5).

You can also index a string that is stored inside a list by chaining indices.

````python
s = 'hello'
print("String:", s)
print("First character:", s[0])
print("Last character:", s[-1])
````

````python
# Indexing a string inside a list
words = ["thrust", "velocity", "altitude"]

# Get the first character of the second word
print("Second word:", words[1])
print("First character of second word:", words[1][0])

# Strings are immutable — this would cause an error:
# s = "hello"
# s[0] = "H"  # TypeError: 'str' object does not support item assignment
````

## 6. Indexing Tuples and Dictionaries

**Tuples** use the same indexing as lists — you access elements by position number using `[]`. The difference is that tuples are **immutable**, so you cannot change their elements (we learned this in Notebook 5).

**Dictionaries** are different — they do not use position numbers. Instead, you access values by their **key** (we learned this in Notebook 7). This is sometimes called "key-based access" rather than "index-based access."

| Feature | List | Tuple | Dictionary |
|---------|------|-------|------------|
| Access by | Position index | Position index | Key name |
| Syntax | `my_list[0]` | `my_tuple[0]` | `my_dict["key"]` |
| Mutable? | Yes | No | Yes |
| Negative index? | Yes | Yes | No |

````python
# Tuple indexing — works the same as lists
coordinates = (10.5, 20.3, 30.1)
print("Tuple:", coordinates)
print("First element:", coordinates[0])
print("Last element:", coordinates[-1])

# Dictionary access — uses keys, not position numbers
sensor_data = {"temperature": 22.5, "pressure": 101.3, "humidity": 45}
print("\nDictionary:", sensor_data)
print("Temperature:", sensor_data["temperature"])
print("Pressure:", sensor_data["pressure"])
````

## 7. Common Pitfalls

Here are the most common mistakes beginners make with indexing:

**1. IndexError — accessing an index that does not exist**
If your list has 5 elements (indices 0–4), trying to access index 5 will crash your program.

**2. Off-by-one errors — forgetting that indexing starts at 0**
The "third element" is at index `2`, not index `3`. This is the most common source of bugs!

**3. TypeError — using a float as an index**
Python requires indices to be integers. Using `2.0` instead of `2` will cause an error.

**4. TypeError — trying to modify an immutable sequence**
You cannot change characters in a string or elements in a tuple by assigning to an index.

````python
# Pitfall 1: IndexError
my_list = [10, 20, 30]
# print(my_list[3])  # IndexError: list index out of range
# Valid indices are 0, 1, 2 (or -1, -2, -3)
print("List length:", len(my_list))
print("Last valid index:", len(my_list) - 1)

# Pitfall 3: TypeError with float index
# print(my_list[1.0])  # TypeError: list indices must be integers
print("Correct way:", my_list[int(1.0)])
````
