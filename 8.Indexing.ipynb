## Python Basics — Indexing

In this notebook, we will explore **indexing**, a fundamental concept in Python for accessing specific elements within sequences like lists, strings, and tuples.

## Learning Goals
By the end of this notebook, students will be able to:
- Understand that Python uses **zero-based indexing**.
- Access individual elements in a list or string using square brackets `[]`.
- Use negative indexing to access elements from the end of a sequence.
- Access elements within nested lists (lists inside lists).
- Recognize that lists cannot be indexed with another list (unlike NumPy arrays).


## 1. Zero-Based Indexing
Python starts counting from 0. The first element is at index 0, the second at index 1, and so on.

**Example:**

```python
# Create a range of numbers and convert to a list
arange = range(10, 15)
alist = list(arange)
print("List:", alist)

# Access the second element (index 1)
print("Element at index 1:", alist[1])
```

## 2. Negative Indexing
Python allows you to access elements from the end of a list using negative numbers. `-1` refers to the last element, `-2` to the second-to-last, etc.


```python
alist = [54, 63, 65, 9, 0, -644, 524]
print("List:", alist)

# Access the last element
print("Last element (-1):", alist[-1])

# Access the third-to-last element
print("Third-to-last element (-3):", alist[-3])
```

## 3. Using Variables as Indices
You can use a variable to specify the index.


```python
idx = 3
print("Element at index", idx, ":", alist[idx])

# Note: You cannot use a list as an index for a standard Python list.
# idx = [3, 4]
# alist[idx]  # This would raise a TypeError. We will learn how to do this with NumPy later.
```

## 4. Nested Indexing
When you have lists inside lists (or other nested structures), you can chain indices to access deeper elements.

**Example:**

```python
alist = [4, 3.2, [9, 4, 6], 32]
print("Nested list:", alist)

# Access the inner list at index 2
inner_list = alist[2]
print("Inner list:", inner_list)

# Access the third element (index 2) of the inner list
element = inner_list[2]
print("Element from inner list:", element)

# Shortcut: Chain the indices
print("Direct access:", alist[2][2])
```

## 5. Indexing Strings
Strings are sequences of characters, so you can index them just like lists.


```python
s = 'hello'
print("String:", s)
print("First character:", s[0])
print("Last character:", s[-1])
```
