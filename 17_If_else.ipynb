## 1. Basic If Statement
The `if` statement executes a block of code only if a specified condition is true.

**Example:**

```python
if True:
    print('This will be printed because the condition is True.')

if False:
    print('This will NOT be printed.')

if 4 == 4:
    print('4 equals 4, so this prints.')
```

## 2. Indentation
Python uses indentation (whitespace) to define blocks of code. All statements inside an `if` block must be indented.


```python
import numpy as np

randnum = np.random.randn()
if randnum > 0:
    print(randnum)
    x = 3
    print("Inside if block: x =", x)

print('Outside if block')  # This runs regardless of the condition
```

## 3. If-Else Statement
The `else` statement allows you to execute a block of code when the `if` condition is false.


```python
r = np.random.randn()
if r > 0:
    print(str(r) + ' is positive')
else:
    print(str(r) + ' is negative (or zero)')
```

## 4. Elif Statement
The `elif` (else if) statement allows you to check multiple conditions sequentially.


```python
caketype = 'chocolate'

if caketype == 'carrot':
    print('Total yumminess!!')
elif caketype == 'chocolate':
    print('Still good')
elif caketype == 'fruit':
    print('Healthy')
else:
    print('Cake type not recognized')
```

## 5. One-Line If Statements
For simple conditions, you can write the `if` statement on a single line.


```python
caketype = 'Carrot'

# Simple one-liner
if caketype[0].lower() == 'c': print('Probably carrot cake')

# Ternary operator (conditional expression)
print('Probably carrot cake') if caketype[0].lower() == 'c' else print('Probably not carrot cake')
```

## 6. Conjunctive Conditionals
You can combine multiple conditions using `and` and `or`.


```python
if 1 < 3 and 2 < 4:
    print('Both conditions are True')

if 10 < 3 or 2 < 4:
    print('At least one condition is True')
```

```python
import numpy as np

def my_dot_product(v1, v2):
    # Check if inputs are numpy arrays
    if not isinstance(v1, np.ndarray) or not isinstance(v2, np.ndarray):
        raise TypeError("Both inputs must be NumPy arrays.")

    # Check if lengths are equal
    if len(v1) != len(v2):
        raise ValueError("Vectors must have the same length.")

    # Compute dot product
    return np.sum(v1 * v2)

# Test the function
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

try:
    result = my_dot_product(v1, v2)
    print("Dot product:", result)

    # Test error handling (uncomment to test)
    # my_dot_product(v1, [1, 2]) # Should raise TypeError
    # my_dot_product(v1, np.array([1, 2])) # Should raise ValueError

except Exception as e:
    print("Error:", e)
```
