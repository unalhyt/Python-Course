## 1. Basic If Statement
The `if` statement executes a block of code only if a specified condition is true.

**Example:**

```python
if True:
    print('This will be printed because the condition is True.')

if False:
    print('This will NOT be printed.')

if 4 == 4:
    print('4 equals 4, so this prints.')
```

## 2. Indentation
Python uses indentation (whitespace) to define blocks of code. All statements inside an `if` block must be indented.


```python
import numpy as np

randnum = np.random.randn()
if randnum > 0:
    print(randnum)
    x = 3
    print("Inside if block: x =", x)

print('Outside if block')  # This runs regardless of the condition
```

## 3. If-Else Statement
The `else` statement allows you to execute a block of code when the `if` condition is false.


```python
r = np.random.randn()
if r > 0:
    print(str(r) + ' is positive')
else:
    print(str(r) + ' is negative (or zero)')
```

## 4. Elif Statement
The `elif` (else if) statement allows you to check multiple conditions sequentially.


```python
caketype = 'chocolate'

if caketype == 'carrot':
    print('Total yumminess!!')
elif caketype == 'chocolate':
    print('Still good')
elif caketype == 'fruit':
    print('Healthy')
else:
    print('Cake type not recognized')
```

## 5. One-Line If Statements
For simple conditions, you can write the `if` statement on a single line.


```python
caketype = 'Carrot'

# Simple one-liner
if caketype[0].lower() == 'c': print('Probably carrot cake')

# Ternary operator (conditional expression)
print('Probably carrot cake') if caketype[0].lower() == 'c' else print('Probably not carrot cake')
```

## 6. Conjunctive Conditionals
You can combine multiple conditions using `and` and `or`.


```python
if 1 < 3 and 2 < 4:
    print('Both conditions are True')

if 10 < 3 or 2 < 4:
    print('At least one condition is True')
```

## 9. Comparison Operators — Reference Table

When writing conditions, you compare values using these operators. They are the building blocks of every `if` statement.

| Operator | Meaning                | Example          | Result   |
|:--------:|:-----------------------|:-----------------|:--------:|
| `==`     | Equal to               | `5 == 5`         | `True`   |
| `!=`     | Not equal to           | `5 != 4`         | `True`   |
| `>`      | Greater than           | `5 > 3`          | `True`   |
| `<`      | Less than              | `3 < 5`          | `True`   |
| `>=`     | Greater than or equal  | `5 >= 5`         | `True`   |
| `<=`     | Less than or equal     | `4 <= 5`         | `True`   |
| `in`     | Membership             | `'a' in 'cat'`   | `True`   |
| `not in` | Not a member           | `5 not in [1,2]` | `True`   |

⚠️ A common mistake: `=` (assignment) is **not** the same as `==` (comparison).

```python
# Comparison operators always return a Boolean (True or False)
print("5 == 5  ->", 5 == 5)        # equal to
print("5 != 4  ->", 5 != 4)        # not equal to
print("3 < 5   ->", 3 < 5)         # less than
print("5 >= 5  ->", 5 >= 5)        # greater or equal

# 'in' works for strings, lists, tuples
fruits = ['apple', 'banana', 'cherry']
print("'apple' in fruits  ->", 'apple' in fruits)
print("'grape' in fruits  ->", 'grape' in fruits)

# Chained comparisons (a math-friendly Python feature)
temperature = 25
print("Is room temp (15 < t < 30)?", 15 < temperature < 30)
```

## 10. Truthiness — When Things Are "True" Without Being `True`

In Python, `if` does not need a Boolean. It checks if a value is **truthy** (treated like `True`) or **falsy** (treated like `False`). Think of it as: *"is there something here?"*

**Falsy values** (treated as `False`):
- `False`
- `0`, `0.0`
- `""` (empty string)
- `[]`, `()`, `{}` (empty list, tuple, dict)
- `None`

Everything else is **truthy**. This is very useful for checking if a list of sensor readings has any data.

```python
# Engineering example: check if we received any sensor readings
readings = []  # empty list — no data yet
if readings:
    print("Processing", len(readings), "readings")
else:
    print("No readings received — skipping analysis")

# Adding values flips the result
readings = [3.14, 2.71]
if readings:
    print("Processing", len(readings), "readings")

# Be careful: 0 is falsy too!
sensor_value = 0
if sensor_value:
    print("Sensor reports a value")
else:
    print("Sensor value is 0 (or missing) — looks falsy")
```

## 11. Nested If Statements

You can place an `if` inside another `if`. This is useful when one condition only matters after a previous one passes — e.g., classifying a temperature.

Often a chain of `elif` is cleaner than nested `if`. Compare the two styles in the example below.

```python
# Engineering example: classify a temperature reading using nested if
temperature_c = 28  # Celsius

if temperature_c >= 0:           # only check warm/hot if not freezing
    if temperature_c < 15:
        print("Cool")
    elif temperature_c < 25:
        print("Warm")
    else:
        print("Hot")
else:
    print("Below freezing")

# The same logic written more cleanly using elif (no nesting needed):
if temperature_c < 0:
    print("Below freezing")
elif temperature_c < 15:
    print("Cool")
elif temperature_c < 25:
    print("Warm")
else:
    print("Hot")
```

## 12. Common Pitfalls

1. **Using `=` instead of `==`** — `=` assigns, `==` compares. `if x = 5:` is a SyntaxError.

2. **Forgetting the colon `:`** — every `if`, `elif`, `else` line ends with `:`.

3. **Wrong indentation** — Python uses indentation to know what belongs inside the `if` block. Mixing tabs and spaces causes errors.

4. **Using `or` incorrectly** — `if x == 1 or 2:` is **always True**, because `2` alone is truthy. Write `if x == 1 or x == 2:` instead.

5. **Comparing floats with `==`** — Floats may not compare exactly because of how computers store them. Use a tolerance: `abs(a - b) < 1e-9`.

The next cell demonstrates pitfalls #4 and #5.

```python
# Pitfall #4 demo — wrong vs right way to test multiple values
x = 7

# WRONG — this is always True!
# Reason: 'or 2' is evaluated as bool(2), which is True
if x == 1 or 2:
    print("Wrong way: triggered (and it always will!)")

# RIGHT — repeat the variable on each side of 'or'
if x == 1 or x == 2:
    print("Right way: triggered")
else:
    print("Right way: x is not 1 or 2")

# Pitfall #5 demo — float comparison
a = 0.1 + 0.2
b = 0.3
print("a == b ?", a == b)                  # False! (floating point quirk)
print("close enough ?", abs(a - b) < 1e-9) # True (the safer test)
```

**Task A — Grade Calculator:** Write a function `grade(score)` that returns:
- `'A'` if score ≥ 90
- `'B'` if score ≥ 80
- `'C'` if score ≥ 70
- `'D'` if score ≥ 60
- `'F'` otherwise

**Task B — Vector Sign Classifier:** Given a NumPy array `v`, print whether it is "All positive", "All negative", or "Mixed". Hint: `np.all(v > 0)` checks if every element is positive.

**Task C — Leap Year Checker:** Write `is_leap(year)` that returns `True` if year is a leap year. A year is a leap year if it is divisible by 4 — *except* if it is divisible by 100, *unless* it is also divisible by 400.

```python
# Solution to Task A — Grade Calculator
def grade(score):
    # elif checks each band only after the previous one fails
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

# Test with several scores
for s in [95, 82, 71, 60, 45]:
    print("Score", s, "->", grade(s))
```

```python
# Solution to Task B — Vector Sign Classifier
import numpy as np

def classify_vector(v):
    if np.all(v > 0):           # every element > 0?
        print("All positive")
    elif np.all(v < 0):         # every element < 0?
        print("All negative")
    else:
        print("Mixed (or contains zero)")

# Test with three vectors
classify_vector(np.array([1, 2, 3]))     # all positive
classify_vector(np.array([-1, -2, -3]))  # all negative
classify_vector(np.array([1, -2, 3]))    # mixed
```

```python
# Solution to Task C — Leap Year Checker
def is_leap(year):
    # Rule: divisible by 400 -> leap
    #       divisible by 100 (but not 400) -> NOT leap
    #       divisible by 4 (but not 100)   -> leap
    #       otherwise -> NOT leap
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

# Test with several known years
for y in [2000, 1900, 2024, 2023, 2100]:
    print(y, "->", is_leap(y))
```
