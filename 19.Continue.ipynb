## Python Basics — Continue & Break

In this notebook, we will explore the `continue` and `break` statements, which allow you to control the flow of loops more precisely.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `continue` to skip the rest of the current iteration and proceed to the next one.
- Use `break` to exit the loop entirely.
- Apply these statements to filter data or stop processing early.


## 1. The `continue` Statement
The `continue` statement skips the remaining code inside the loop for the current iteration and jumps to the next iteration.

**Example:**

```python
# Print only odd numbers
for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)
```

## 2. The `break` Statement
The `break` statement terminates the loop immediately, regardless of the loop condition.

**Example:**

```python
# Stop the loop when i reaches 5
for i in range(10):
    if i == 5:
        break  # Exit the loop
    print(i)
```

## 3. Practice Task: Filtering Vowels
**Exercise:**
Given the string `text`, print only the consonants (non-vowels).
- Use `continue` to skip vowels.
- Use `end=''` in `print()` to print on the same line.


```python
text = 'HellO, my name is MikE'

print("Original text:", text)
print("Consonants only: ", end='')

for char in text:
    # Check if the character is a vowel (case-insensitive)
    if char.lower() in 'aeiou':
        continue
    print(char, end='')

print()  # New line at the end
```

## 5. The `else` Clause on Loops

Python has a less-known feature: a `for` (or `while`) loop can have an `else` block. The `else` runs **only if the loop finished normally without `break`**.

This is handy for searches: "did we find what we were looking for, or did we go through the whole list?"

```python
# Loop's else branch — runs only if no break occurred
target = 7

# Search 1: target is in the list -> break fires -> else SKIPPED
print("Search 1 (target found):")
for x in [1, 3, 5, 7, 9]:
    if x == target:
        print("Found", target)
        break
else:
    print("Not found")          # not printed

# Search 2: target NOT in list -> loop completes -> else RUNS
print("\nSearch 2 (target missing):")
for x in [1, 3, 5, 9]:
    if x == target:
        print("Found", target)
        break
else:
    print("Not found")          # this prints
```

## 6. `continue` vs `break` — Side-by-Side Comparison

| Statement   | What it does                                | Where it goes next         |
|:-----------:|:--------------------------------------------|:---------------------------|
| `continue`  | Skips the rest of the current iteration     | Back to the loop condition |
| `break`     | Exits the loop entirely                     | The first line **after** the loop |
| `pass`      | Does nothing (placeholder)                  | The next line in the loop  |
| `return`    | Exits the loop **and** the enclosing function | Caller of the function     |

Think of `continue` like skipping a song in a playlist (next song plays). Think of `break` like turning off the music (playlist stops).

## 7. Engineering Use Cases

`continue` and `break` are not just classroom curiosities — they are very common in real engineering code:

- **`break`** — *stop early when a goal is reached*. Examples: an iterative solver hitting a tolerance, finding a target value in a sensor log, exiting an optimization once converged.
- **`continue`** — *skip invalid or unwanted items*. Examples: skipping `NaN` readings, ignoring outliers, filtering out comments in a data file.

The next cell demonstrates both patterns.

```python
import numpy as np

# Engineering example #1: skip invalid (NaN) sensor readings using continue
readings = [3.2, np.nan, 4.1, np.nan, 5.5, 4.7]
total = 0
count = 0
for r in readings:
    if np.isnan(r):
        continue            # skip missing data
    total += r
    count += 1

print("Average of valid readings:", total / count)

# Engineering example #2: stop early when convergence is reached, using break
errors = [0.5, 0.2, 0.08, 0.03, 0.005, 0.001]   # iterative solver errors
tolerance = 0.01
for i, err in enumerate(errors):
    if err < tolerance:
        print("Converged at iteration", i, "with error", err)
        break               # no need to keep iterating
```

## 8. Common Pitfalls

1. **`break` only exits the innermost loop** — In nested loops, it does not jump out of the outer loop. You'd need a flag variable or wrap the outer loop in a function and `return`.

2. **`continue` in a `while` loop with no counter update** — leads to an **infinite loop**. Always update the loop counter *before* `continue` (or be sure the condition will change).

3. **Forgetting that `else` on a loop only runs without `break`** — If the loop ever hits `break`, the `else` block is skipped.

4. **Using `break` when a `return` is cleaner** — Inside a function, `return` exits both the loop and the function. Often simpler than `break` + flag.

```python
# Pitfall #1 — break only exits the INNER loop in nested loops
print("Pitfall #1: break only exits the inner loop")
for i in range(3):              # outer loop
    for j in range(3):          # inner loop
        if j == 1:
            break               # only exits the j-loop
        print("i=", i, "j=", j)

# Pitfall #2 — continue without updating a counter -> infinite loop
print("\nPitfall #2: safe continue (counter is updated)")
i = 0
while i < 5:
    i += 1                      # update i FIRST so continue is safe
    if i == 3:
        continue                # skip 3
    print("i =", i)
```

## 9. More Practice Tasks

**Task A — Find First Negative:** Write `first_negative(values)` that prints and returns the first negative number in a list. If none found, print "No negative numbers found". Use `break` (or `return`).

**Task B — Sum Only Positives:** Write `sum_positives(values)` that returns the sum of only the positive numbers in a list. Use `continue` to skip negatives.

**Task C — Search:** Write `search(values, target)` that prints the index of `target` in `values`, or "not in list" if absent. Stop searching as soon as you find it.

```python
# Solution to Task A — Find First Negative Number
def first_negative(values):
    for x in values:
        if x < 0:
            print("First negative is:", x)
            return x        # return immediately — like 'break' but with a value
    print("No negative numbers found")
    return None             # if loop finishes naturally

# Test cases
first_negative([3, 7, -2, 4, -5])      # finds -2
first_negative([1, 2, 3])              # none
```

```python
# Solution to Task B — Sum Only Positive Numbers (skip negatives)
def sum_positives(values):
    total = 0
    for x in values:
        if x < 0:
            continue        # skip negative values
        total += x          # add only positives (and zero)
    return total

# Test
data = [3, -1, 4, -2, 5, -8, 9]
print("Positives sum =", sum_positives(data))   # 3 + 4 + 5 + 9 = 21
```

```python
# Solution to Task C — Search for a Value in a List
def search(values, target):
    # Loop with index using enumerate
    for i, v in enumerate(values):
        if v == target:
            print("Found", target, "at index", i)
            return i        # found — return immediately (acts like break)
    print(target, "not in list")
    return -1               # not found — return -1 by convention

# Test cases
data = [10, 20, 30, 40, 50]
search(data, 30)            # found
search(data, 99)            # not found
```
