## Python Basics — Continue & Break

In this notebook, we will explore the `continue` and `break` statements, which allow you to control the flow of loops more precisely.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `continue` to skip the rest of the current iteration and proceed to the next one.
- Use `break` to exit the loop entirely.
- Apply these statements to filter data or stop processing early.


## 1. The `continue` Statement
The `continue` statement skips the remaining code inside the loop for the current iteration and jumps to the next iteration.

**Example:**

```python
# Print only odd numbers
for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)
```

## 2. The `break` Statement
The `break` statement terminates the loop immediately, regardless of the loop condition.

**Example:**

```python
# Stop the loop when i reaches 5
for i in range(10):
    if i == 5:
        break  # Exit the loop
    print(i)
```

## 3. Practice Task: Filtering Vowels
**Exercise:**
Given the string `text`, print only the consonants (non-vowels).
- Use `continue` to skip vowels.
- Use `end=''` in `print()` to print on the same line.


```python
text = 'HellO, my name is MikE'

print("Original text:", text)
print("Consonants only: ", end='')

for char in text:
    # Check if the character is a vowel (case-insensitive)
    if char.lower() in 'aeiou':
        continue
    print(char, end='')

print()  # New line at the end
```
