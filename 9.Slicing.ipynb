## Python Basics — Slicing

In this notebook, we will explore **slicing**, a powerful technique in Python for extracting subsequences (slices) from lists, strings, and other sequences.

- Use default values for `start`, `stop`, and `step`.
- Reverse a sequence using slicing with a negative step.
- Apply slicing to strings and lists.


## 1. Basic Slicing
Slicing allows you to get a **range of items** from a sequence. The syntax is `sequence[start:stop]`.

**Think of it like this:** Imagine you have a long rope with numbered marks. Slicing is like cutting out a section of that rope — you choose where to start cutting and where to stop.

**Important:** The element at the `stop` index is **not** included in the slice. This is the same "exclusive stop" rule you saw with `range()` in earlier notebooks.

We learned how to access **single elements** with indexing in Notebook 8. Slicing lets us grab **multiple elements at once**.

Here is a visual diagram:

```
List:    [  5,   6,   7,   8,   9,  10 ]
Index:      0    1    2    3    4    5
Slice [1:4] gives us: [6, 7, 8]  (indices 1, 2, 3 — stop index 4 is excluded)
```

````python
# Create a list of numbers from 5 to 10
alist = list(range(5, 11))
print("Original list:", alist)

# Get the first two elements (indices 0 and 1)
print("Slice [0:2]:", alist[0:2])

# Get elements from index 2 to the end
print("Slice [2:]:", alist[2:])

# Get elements from the beginning up to index 4 (exclusive)
print("Slice [:4]:", alist[:4])
````

````python
# Slicing works on any sequence — here with chemical elements
elements = ["H", "He", "Li", "Be", "B", "C", "N", "O"]
print("All elements:", elements)

# Get the first 3 elements
print("First 3 elements [:3]:", elements[:3])

# Get elements from index 2 to 5 (exclusive)
print("Elements [2:5]:", elements[2:5])

# Get from index 4 to the end
print("From index 4 onwards [4:]:", elements[4:])
````

## 2. Step Slicing
You can specify a third parameter, `step`, to skip elements. The full syntax is `sequence[start:stop:step]`.

**Think of it like this:** Imagine walking along a row of mailboxes. With `step=1` you check every mailbox. With `step=2` you check every other mailbox. With `step=3` you check every third one.

```
List:       [  5,   6,   7,   8,   9,  10 ]
Index:         0    1    2    3    4    5
[::2] picks:   5         7         9        (every 2nd element)
[::3] picks:   5              8              (every 3rd element)
```

````python
# Get elements from index 0 to 5, skipping every 2nd element
print("Slice [0:5:2]:", alist[0:5:2])

# Get every 3rd element from the entire list
print("Slice [::3]:", alist[::3])
````

````python
# Real-world use: downsample sensor data by taking every 5th reading
sensor_readings = [10, 12, 11, 13, 14, 15, 13, 12, 16, 18, 17, 15, 14, 13, 12]
downsampled = sensor_readings[::5]  # every 5th value
print("All readings:", sensor_readings)
print("Every 5th reading:", downsampled)

# Step slicing with strings — get every 3rd letter
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
print("Every 3rd letter:", alphabet[::3])
````

## 3. Reversing with Slicing
Using a negative step value allows you to traverse the sequence **backwards**. The most common pattern is `[::-1]` which reverses the entire sequence.

**Think of it like this:** Imagine reading a book from the last page to the first page — that is what a negative step does.

When using a negative step, `start` should be a **higher** index than `stop` (you are counting backwards).

```
List:       [  5,   6,   7,   8,   9,  10 ]
Index:         0    1    2    3    4    5
[::-1] →    [ 10,   9,   8,   7,   6,   5 ]   (reversed)
[4:1:-1] →  [  9,   8,   7 ]                   (from index 4 down to 2)
```

````python
# Reverse the entire list
print("Reversed list [::-1]:", alist[::-1])

# Reverse a portion of the list
print("Reversed slice [4:1:-1]:", alist[4:1:-1])
````

````python
# Practical use: check if a string is a palindrome (reads same forwards and backwards)
word = "racecar"
is_palindrome = word == word[::-1]  # compare original to reversed
print(f"'{word}' is a palindrome:", is_palindrome)

word2 = "python"
is_palindrome2 = word2 == word2[::-1]
print(f"'{word2}' is a palindrome:", is_palindrome2)

# Reverse a string with step -2 (every other character, backwards)
print("Every other char reversed:", "ABCDEFGH"[::-2])
````

## 4. Using Variables for Slicing
You can use variables to define the start, stop, and step values, making your code more dynamic.

This is especially useful when you do not know the exact size of your data. You can use `len()` to calculate slice boundaries at runtime.

A common pattern is splitting data into two halves:
```
data = [1, 2, 3, 4, 5, 6]
mid = len(data) // 2           # mid = 3
first_half  = data[:mid]       # [1, 2, 3]
second_half = data[mid:]       # [4, 5, 6]
```

````python
start = 1
stop = 4
skip = 2

print("Slice with variables:", alist[start:stop:skip])
````

````python
# Split flight altitude data into first half and second half
altitudes = [1000, 2500, 5000, 8000, 10000, 12000, 11000, 9000]
mid = len(altitudes) // 2  # find the middle index

first_half = altitudes[:mid]   # climb phase
second_half = altitudes[mid:]  # descent phase
print("All altitudes:", altitudes)
print("First half (climb):", first_half)
print("Second half (descent):", second_half)

# Get the last 3 readings using negative index with len()
last_three = altitudes[len(altitudes) - 3:]  # same as altitudes[-3:]
print("Last 3 readings:", last_three)
````

## 5. Slicing Creates a Copy
An important fact: when you slice a list, Python creates a **new list**. Changing the slice does **not** change the original.

**Think of it like this:** Slicing is like photocopying pages from a book. You can write on the photocopy without affecting the original book.

This is different from simple assignment (`b = a`), where both variables point to the **same** list. We will cover this in more detail in Notebook 15 (Variable Scopes).

````python
# Slicing creates a COPY — changes to the slice do not affect the original
original = [10, 20, 30, 40, 50]
my_slice = original[1:4]  # [20, 30, 40]

my_slice[0] = 999  # modify the slice
print("Modified slice:", my_slice)
print("Original is unchanged:", original)

# But simple assignment does NOT copy — both names point to the same list
reference = original
reference[0] = 777
print("Reference changed:", reference)
print("Original ALSO changed:", original)
````

## 6. Out-of-Bounds Slicing
Unlike **indexing** (Notebook 8), slicing does **not** raise an error when you go past the end of a sequence. Python simply returns whatever elements exist in that range.

This is a key difference:
| Operation | Out of bounds | Result |
|-----------|--------------|--------|
| `my_list[100]` | **IndexError** | Crash! |
| `my_list[2:100]` | No error | Returns elements from index 2 to the end |
| `my_list[100:200]` | No error | Returns an empty list `[]` |

````python
# Out-of-bounds slicing does NOT cause an error
short_list = [1, 2, 3]

# Stop index beyond the list — Python just returns what exists
print("short_list[1:100]:", short_list[1:100])

# Both start and stop beyond the list — returns empty list
print("short_list[50:100]:", short_list[50:100])

# Negative out-of-bounds — also safe
print("short_list[-100:2]:", short_list[-100:2])

# Compare: indexing WOULD crash → short_list[100] gives IndexError
````

## 7. Common Pitfalls

**1. Off-by-one error — forgetting that `stop` is exclusive**
```python
data = [10, 20, 30, 40, 50]
data[1:3]   # gives [20, 30], NOT [20, 30, 40]
data[1:4]   # gives [20, 30, 40] ← this is what you probably wanted
```

**2. Wrong direction with negative step**
```python
data[1:4:-1]   # gives [] (empty!) — you cannot go from 1 to 4 backwards
data[4:1:-1]   # gives [50, 40, 30] ← correct direction
```

**3. Using `step=0`**
```python
data[::0]   # ValueError: slice step cannot be zero
```

**4. Thinking a slice modifies the original list**
```python
subset = data[1:3]
subset[0] = 999   # only changes subset, NOT data (see Section 5)
```

**5. Confusing slicing with indexing**
```python
data[2]     # returns a single element: 30
data[2:3]   # returns a list with one element: [30]
```
