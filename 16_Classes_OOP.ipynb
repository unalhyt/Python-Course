## Python Basics — Classes & Object-Oriented Programming (OOP)

In this notebook, we will explore **Object-Oriented Programming (OOP)** in Python. We will learn how to define **classes**, create **objects** (instances), and use **methods** to interact with them.

- Create multiple instances of a class.


## 1. Defining a Class
A class is a blueprint for creating objects. It defines the attributes (data) and methods (functions) that the objects will have.

**Example:**

```python
class Experiment(object):
    # Constructor method: Initializes the object
    def __init__(self, eType, dataN, status='ongoing'):
        # self refers to the instance being created
        self.expType = eType      # Attribute: Type of experiment
        self.dataCount = dataN    # Attribute: Number of data points
        self.expStatus = status   # Attribute: Status (default is 'ongoing')

    # Method to increase the data count
    def increaseN(self, number):
        self.dataCount += number

    # Method to change the experiment status
    def changeExpStatus(self, status):
        self.expStatus = status

    # Special method for string representation
    def __str__(self):
        return 'Experiment type: ' + self.expType + ' has ' + str(self.dataCount) + ' data points.'
```

## 2. Creating Objects (Instances)
To create an object from a class, you call the class name as if it were a function, passing the required arguments to the `__init__` method.


```python
# Create a new experiment instance
myexperiment = Experiment('Brain', 120)

# Print the object (uses __str__)
print(myexperiment)
```

## 3. Accessing Attributes and Methods
You can access attributes and call methods using the dot `.` notation.


```python
# Access an attribute
print("Initial data count:", myexperiment.dataCount)

# Call a method to modify the object
myexperiment.increaseN(10)
print("Updated data count:", myexperiment.dataCount)

# Change status
myexperiment.changeExpStatus('finished')
print("New status:", myexperiment.expStatus)

# Note: Methods must be called on an instance.
# changeExpStatus('finished') # This would raise a NameError
```

## 4. Practice Task: Board Game Class
**Exercise:**
Create a class `BoardGame` to store information about your board game hobby.
The class must contain:
- Attributes:
    - Name of the board game
    - Number of times played
    - Number of times won
- Methods:
    - Update number of plays and wins
    - Print the outcome of the last game (win/lose)

Create at least two instances to test your class.


```python
class BoardGame:
    def __init__(self, name, plays=0, wins=0):
        self.name = name
        self.plays = plays
        self.wins = wins
        self.last_result = None # To store the result of the last game

    def record_game(self, won):
        self.plays += 1
        if won:
            self.wins += 1
            self.last_result = "Win"
        else:
            self.last_result = "Loss"

    def __str__(self):
        return f"{self.name}: Played {self.plays} times, Won {self.wins} times."

# Create instances
game1 = BoardGame("Catan")
game2 = BoardGame("Chess", plays=10, wins=5)

# Play games
game1.record_game(won=True)
print(game1)
print("Last game result for Catan:", game1.last_result)

game2.record_game(won=False)
print(game2)
print("Last game result for Chess:", game2.last_result)
```
