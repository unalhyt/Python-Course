## Python Basics — Single-Line Loops (List Comprehensions)

In this notebook, we will explore **list comprehensions**, a concise way to create lists and perform operations in a single line of code. This is a powerful feature of Python that can make your code more readable and efficient.

## Learning Goals
By the end of this notebook, students will be able to:
- Write single-line `for` loops using list comprehensions.
- Combine `if` statements with list comprehensions for filtering.
- Use `if-else` logic within list comprehensions.
- Convert traditional loops into list comprehensions.
- Understand when to use list comprehensions vs. traditional loops.


## 1. Basic List Comprehension
A list comprehension consists of brackets `[]` containing an expression followed by a `for` clause.

**Example:**

````python
# Traditional loop
for i in range(10):
    print(i**2)

# List comprehension (creates a list of results)
squares = [i**2 for i in range(10)]
print("Squares:", squares)
````

## 2. Filtering with If
You can add an `if` clause to filter elements.


````python
# Only include squares where i > 5
filtered_squares = [i**2 - i**(1/2) for i in range(10) if i > 5]
print("Filtered squares:", filtered_squares)
````

## 3. If-Else in List Comprehension
You can use `if-else` logic to conditionally modify elements. Note the syntax: `value_if_true if condition else value_if_false`.


````python
# Create a list where odd numbers are replaced by 'odd'
newlist = ['odd' if i % 2 == 1 else i for i in range(10)]
print("Mixed list:", newlist)
````

## 4. Practice Task: Extracting Initials
**Exercise:**
Convert the following loop into a single-line list comprehension that extracts the first letter of each word.

```python
text = ['Promising', 'Yves', 'that', 'home', 'on', 'Nobb']
for word in text:
    print(word[0])
```


````python
text = ['Promising', 'Yves', 'that', 'home', 'on', 'Nobb']

# Single-line version (creates a list of first letters)
initials = [word[0] for word in text]
print("Initials:", initials)

# To print each one (side effect), you can do:
[print(word[0]) for word in text];  # The semicolon suppresses the output of the list itself
````

## 5. Practice Task: Recreating Linspace
**Exercise:**
Rewrite the creation of `x1` and `y1` using list comprehensions instead of NumPy functions.
- `x1` is `np.linspace(-3, 3, 11)` (11 points from -3 to 3).
- `y1` is `x1**2`.

Hint: The step size for 11 points from -3 to 3 is `(3 - (-3)) / (11 - 1) = 0.6`.


````python
import numpy as np

# Original NumPy version
x1 = np.linspace(-3, 3, 11)
y1 = x1**2

print("NumPy x1:", x1)
print("NumPy y1:", y1)

# List comprehension version
# We need 11 points from -3 to 3. Step size is 0.6.
# Range -5 to 6 gives 11 integers. Divide by (5/3) is approx 1.66, which is 1/0.6.
# Let's use a more direct calculation: start + i * step
step = (3 - (-3)) / (11 - 1)
x2 = [-3 + i * step for i in range(11)]
y2 = [val**2 for val in x2]

print("\nList Comp x2:", x2)
print("List Comp y2:", y2)

# Verify they are close
print("\nAre they close?", np.allclose(x1, x2))
````
