## 1. Old-Style Formatting (%)
The `%` operator allows you to insert values into a string using placeholders like `%s` (string), `%d` (integer), and `%f` (float).

**Example:**

```python
num = 45.3
sng = 'cellar doors'

print('I would like to have %g %s' % (num, sng))  # %g: general number format
print('I would like to have %.2f %s' % (num, sng))  # %.2f: 2 decimal places
```

## 2. The `.format()` Method
The `.format()` method provides more control and readability. You use `{}` as placeholders.


```python
print('I would like to have {} {}'.format(num, sng))
print('I would like to have {n} {s}'.format(n=num, s=sng))  # Named arguments
```

## 3. f-Strings (Recommended)
f-strings are the most modern way to format strings. You prefix the string with `f` and put expressions inside `{}`.


```python
print(f'I would like to have {num} {sng}')
print(f'Calculation: {num / 5**3}')  # You can do math inside {}
```

## 4. Practice Task: Alphabet Sentences
**Exercise:**
Print sentences for each letter of the alphabet:
- "a is the 1st letter of the alphabet."
- "b is the 2nd letter of the alphabet."
- ...
- "z is the 26th letter of the alphabet."

Handle the ordinal suffixes (st, nd, rd, th) correctly.


```python
import string

letters = string.ascii_lowercase
print("Letters:", letters)

for i in range(len(letters)):
    # Determine the ordinal suffix
    # 1st, 2nd, 3rd, 21st, 22nd, 23rd...
    # 11th, 12th, 13th are exceptions (use 'th')

    idx = i + 1  # 1-based index

    if idx % 10 == 1 and idx != 11:
        ordind = 'st'
    elif idx % 10 == 2 and idx != 12:
        ordind = 'nd'
    elif idx % 10 == 3 and idx != 13:
        ordind = 'rd'
    else:
        ordind = 'th'

    # Print using f-string
    print(f'{letters[i]} is the {idx}{ordind} letter of the alphabet.')
```
