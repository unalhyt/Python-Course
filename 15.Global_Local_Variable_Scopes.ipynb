## Python Basics — Global vs. Local Variable Scopes

In this notebook, we will explore the concept of **variable scope** in Python. Understanding the difference between global and local variables is crucial for writing clean and bug-free code.

## 1. Local Variables
Variables created inside a function are **local** to that function. They cannot be accessed from outside the function.

Think of it like notes on a classroom whiteboard: when the class (function) ends, the whiteboard is erased. The notes (local variables) only exist during the class. Other classes cannot see what was written.

**Example:**

````python
def funfun():
    x = 3  # Local variable
    y = 4  # Local variable
    print("Inside funfun, x =", x)

# Call the function
funfun()

# Attempting to access x outside the function will raise a NameError
# print(x)
````

Two different functions can use the **same variable name** without any conflict, because each function has its own local scope:

````python
# Two functions using the same variable name — no conflict!
def experiment_a():
    result = 42  # Local to experiment_a
    print("experiment_a result:", result)

def experiment_b():
    result = 99  # Local to experiment_b (different variable!)
    print("experiment_b result:", result)

experiment_a()
experiment_b()
# Each function has its own separate 'result' variable
````

## 2. Global Variables
Variables created outside of any function are **global**. They can be accessed from anywhere in the code, including inside functions (unless shadowed by a local variable).

> **Caution:** While global variables are convenient for small scripts, overusing them in larger programs makes code harder to understand and debug. It becomes difficult to track where and when they change. Prefer passing values as function arguments when possible (as we learned in Notebook 14).

````python
x = 9  # Global variable

def print_global():
    print("Inside print_global, x =", x)

print_global()
print("Outside print_global, x =", x)
````

## 3. Scope Rules
1.  **Local Scope:** Variables defined inside a function are local.
2.  **Global Scope:** Variables defined outside functions are global.
3.  **Precedence:** If a local variable has the same name as a global variable, the local variable takes precedence within the function.

**Example:**

````python
z = 3  # Global variable

def fz():
    print("Inside fz, z =", z)  # Accesses global z

fz()

# If we define a local z inside fz, it would shadow the global z.
def fz_local():
    z = 10  # Local variable
    print("Inside fz_local, z =", z)

fz_local()
print("Outside, z is still", z)
````

### Scope Rules Summary Table

| Situation | What Happens |
|---|---|
| Variable defined **outside** all functions | It is **global** — accessible everywhere |
| Variable defined **inside** a function | It is **local** — only accessible inside that function |
| Same name used globally and locally | The **local** version takes precedence inside the function |
| Reading a global variable inside a function | Works fine — Python looks up to the global scope |
| Assigning to a variable inside a function | Creates a **new local** variable (does NOT change the global) |

## 3.5 The `global` Keyword

What if you actually want to **modify** a global variable from inside a function? You need to use the `global` keyword to tell Python "I want to change the global version, not create a new local one."

> **Warning:** Using `global` is generally discouraged in larger programs. It makes code harder to follow because any function could change the variable at any time. Use it sparingly and only when necessary.

**Example:**

````python
counter = 0  # Global variable

def increment_counter():
    global counter  # Tell Python we want the global version
    counter = counter + 1
    print("Counter inside function:", counter)

print("Counter before:", counter)
increment_counter()
increment_counter()
increment_counter()
print("Counter after:", counter)  # Global was changed!
````

## 3.6 Variable Lifetime

The **lifetime** of a variable is how long it exists in memory:

- **Global variables** are created when the program starts and exist until the program ends.
- **Local variables** are created when the function is **called** and destroyed when the function **returns**. Each time you call the function, fresh local variables are created.

This means that local variables do not "remember" their values between function calls:

````python
# Local variables are created fresh each time
def count_up():
    n = 0       # n is created fresh each call
    n = n + 1   # Increment to 1
    print("n =", n)

# Each call starts with n = 0 again
count_up()  # n = 1
count_up()  # n = 1 (not 2! — n was destroyed and recreated)
count_up()  # n = 1
````

## 4. Practice Task: Coin Flip Simulation
**Exercise:**
Write a function `coinflip(N)` that simulates flipping a coin `N` times and reports the percentage of heads.
- Use `1` for heads and `0` for tails.
- Calculate the average to find the proportion of heads.


````python
import numpy as np

def coinflip(N):
    # Generate N random numbers from a normal distribution
    # Convert to boolean (True/False) based on > 0
    # True acts as 1 (Heads), False acts as 0 (Tails)
    flips = np.random.randn(N) > 0

    # Calculate the average (proportion of True values)
    pCoinFlips = np.average(flips)

    print(str(N) + ' coin flips had ' + str(pCoinFlips * 100) + '% heads.')

# Test the function
coinflip(100)
coinflip(1000)
````

## 5. Common Pitfalls

1. **`UnboundLocalError` — the most confusing scope bug:**
   - If you **read** a global variable and then try to **assign** to it in the same function (without `global`), Python gets confused. It sees the assignment and treats the variable as local, but you tried to read it before it was assigned.
   ```python
   total = 100
   def update():
       total = total + 1  # ERROR! Python thinks total is local
   ```
   Fix: Use `global total` at the top of the function, or pass `total` as an argument.

2. **Accidentally shadowing built-in names:**
   - If you name a variable `list`, `sum`, `max`, `print`, etc., you overwrite the built-in function!
   ```python
   list = [1, 2, 3]  # Now list() no longer works as a function!
   ```
   Fix: Use descriptive names like `my_list`, `data_list`, `measurements`.

3. **Relying on global state makes debugging harder:**
   - When a function depends on global variables, it is hard to test and predict. Prefer passing values as function arguments — this makes your functions self-contained and easier to understand.

4. **Forgetting that assignment creates a new local variable:**
   - Writing `x = 5` inside a function creates a **new local** `x`. It does NOT change a global `x`. This surprises many beginners.

````python
# Demonstrating the UnboundLocalError
total = 100

def bad_update():
    # Python sees 'total = ...' and thinks total is local
    # But we try to read it before assigning — ERROR!
    try:
        total = total + 1
    except UnboundLocalError as e:
        print("Error:", e)
        print("Fix: Add 'global total' or pass total as argument.")

bad_update()
````

## 7. Task Solutions

````python
# Task 1 Solution
temperature = 25

def set_temperature():
    temperature = 100       # This creates a LOCAL variable
    print("Inside:", temperature)  # Prints 100

set_temperature()
print("Outside:", temperature)  # Prints 25 — global was NOT changed
````

````python
# Task 2 Solution
num_calls = 0  # Global counter

def call_counter():
    global num_calls       # We need to modify the global
    num_calls = num_calls + 1
    print(f"This function has been called {num_calls} times.")

# Test it
call_counter()  # 1 time
call_counter()  # 2 times
call_counter()  # 3 times
````

````python
# Task 3 Solution
a = 10
b = 20

def outer():
    a = 50                   # Local a (shadows global a)
    print("outer a:", a)     # Prints 50 (local)
    print("outer b:", b)     # Prints 20 (reads global b)

outer()
print("global a:", a)        # Prints 10 (global was NOT changed)
print("global b:", b)        # Prints 20
````
