## Python Basics — Lists

In this notebook, we will explore **lists**, one of the most versatile and commonly used data structures in Python. Lists allow you to store collections of items, which can be of different types.

## Learning Goals
By the end of this notebook, students will be able to:
- Create lists containing numbers, strings, and mixed data types.
- Understand that lists can contain other lists (nested lists).
- Check for the existence of an item in a list using `in` and `not in`.
- Perform basic list operations like concatenation (`+`) and repetition (`*`).
- Use list methods such as `.append()`, `.sort()`, and `.count()`.


## 1. Creating Lists
Lists are created using square brackets `[]`. Elements inside a list are separated by commas.

**Example:**

```python
# A list of integers
aList = [0, 1, 2, 3, 4, 5]
print("List of numbers:", aList)

# A list of strings
stringList = ['hi', 'my', 'name', 'is', 'Ünal']
print("List of strings:", stringList)

# A list with mixed data types (int, float, str)
mixList = [3.5, 'hello', '7', 7, 7.0]
print("Mixed list:", mixList)
```

## 2. Nested Lists (Lists within Lists)
A list can contain other lists as elements. This is useful for representing matrices or structured data.


```python
listList = [
    3,              # Integer
    [3, 'four', 4], # Another list inside
    'five',         # String
    896             # Integer
]
print("Nested list:", listList)
```

## 3. Checking for Membership
You can check if a specific value exists in a list using the `in` keyword. It returns `True` or `False`.


```python
aList = [0, 1, 2, 3, 4, 5]

# Is 4 in the list?
print("Is 4 in aList?", 4 in aList)

# Is 4.5 in the list?
print("Is 4.5 in aList?", 4.5 in aList)

# Is 4.5 NOT in the list?
print("Is 4.5 not in aList?", 4.5 not in aList)
```

## 4. List Operations
Lists support some mathematical-like operations:
- `+` concatenates (joins) two lists.
- `*` repeats a list multiple times.


```python
print("Original aList:", aList)
print("Original stringList:", stringList)

# Concatenation: Join two lists
combined_list = aList + stringList
print("Combined list:", combined_list)

# Repetition: Repeat the list 3 times
repeated_list = aList * 3
print("Repeated list:", repeated_list)
```

## 5. List Methods
Lists have built-in **methods** (functions that belong to an object) that allow you to modify them or get information.

Common methods:
- `.append(item)`: Adds an item to the end of the list.
- `.sort()`: Sorts the list in ascending order (modifies the original list).
- `.count(item)`: Returns the number of times an item appears in the list.


```python
print("Before append:", aList)

# Add -100 to the end of the list
aList.append(-100)
print("After append:", aList)

# Sort the list
aList.sort()
print("After sort:", aList)

# Tip: In an IDE, typing 'aList.' usually shows a dropdown of available methods.
```
