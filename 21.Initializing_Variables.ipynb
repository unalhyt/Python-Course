## Python Basics — Initializing Variables

In this notebook, we will explore the importance of **initializing variables** before using them. Proper initialization is crucial for performance and avoiding errors.

## Learning Goals
By the end of this notebook, students will be able to:
- Understand why variables must be initialized before use.
- Initialize NumPy arrays using `np.zeros()` and `np.full()`.
- Compare the performance of different initialization methods (list append vs. NumPy append vs. pre-allocation).
- Measure execution time using `time.perf_counter()`.


## 1. Why Initialize?
Attempting to assign values to an undefined variable will result in a `NameError`.

**Example:**

```python
# This will raise a NameError because r is not defined
try:
    for i in range(10):
        r[i] = i**2
    print(r)
except NameError as e:
    print("Error:", e)
```

## 2. Initializing with NumPy
It is common practice to initialize arrays with zeros or a specific value before populating them. This is called **pre-allocation**.


```python
import numpy as np

n = 5  # Soft coding parameter
r = np.zeros(n)  # Initialize with zeros

for i in range(n):
    r[i] = i**2

print("Initialized array:", r)
```

## 3. Other Initialization Methods
You can also initialize arrays with a specific value using `np.full()`.


```python
n = 5
r = np.full(n, 5.5)  # Initialize with 5.5
print("Full array:", r)
```

## 4. Dynamic Growth (Append)
If you don't know the size of the array in advance, you can start with an empty list or array and append elements. However, this is generally slower than pre-allocation.


```python
# Using list append
r_list = []
for i in range(n):
    r_list.append(i**2)
print("List append:", r_list)

# Using NumPy append (less efficient)
r_np = np.array([])
for i in range(n):
    r_np = np.append(r_np, i**2)
print("NumPy append:", r_np)
```

## 5. Practice Task: Performance Comparison
**Exercise:**
Compare the execution time of three methods for creating an array of squares:
1.  List append
2.  NumPy append
3.  NumPy initialization (pre-allocation)

Use `time.perf_counter()` to measure the time.


```python
import time

N = 10000
t = [0] * 3  # To store times

# 1. List append
tic = time.perf_counter()
r = []
for i in range(N):
    r.append(i**2)
t[0] = (time.perf_counter() - tic) * 1000  # Convert to milliseconds

# 2. NumPy append
tic = time.perf_counter()
r = np.array([])
for i in range(N):
    r = np.append(r, i**2)
t[1] = (time.perf_counter() - tic) * 1000

# 3. NumPy initialization
tic = time.perf_counter()
r = np.zeros(N)
for i in range(N):
    r[i] = i**2
t[2] = (time.perf_counter() - tic) * 1000

print("Execution times (ms):")
print(f"List append: {t[0]:.4f}")
print(f"NumPy append: {t[1]:.4f}")
print(f"NumPy init:  {t[2]:.4f}")

# Note: NumPy initialization is usually the fastest, followed by list append. NumPy append is very slow in loops.
```
