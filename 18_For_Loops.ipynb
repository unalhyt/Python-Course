## 1. Basic For Loop
A `for` loop iterates over a sequence (like a list, tuple, or range) and executes the code block for each item.

**Example:**

```python
# Iterate 5 times (0 to 4)
for i in range(5):
    print(i)
```

## 2. Using `range()`
The `range()` function generates a sequence of numbers. It can take start, stop, and step arguments.


```python
# Iterate from 4 to 9 (10 is exclusive)
for j in range(4, 10):
    print(j, j**2)
```

## 3. Iterating Over Arrays
You can iterate over the elements of a NumPy array or list.


```python
import numpy as np

numbers = np.linspace(3, 17, 5)  # 5 numbers from 3 to 17
print("Numbers:", numbers)

for i in range(len(numbers)):
    x = numbers[i]
    print('Iteration ' + str(i+1) + ' has the value of ' + str(x))
```

## 4. Soft Coding vs. Hard Coding
- **Hard Coding:** Using fixed values directly in the code (e.g., `range(100)`). This makes the code less flexible and harder to maintain.
- **Soft Coding:** Using variables to define parameters (e.g., `range(nIterations)`). This makes the code more adaptable.

**Example:**

```python
# Soft coding example
nIterations = 5

# ... 100s of lines of code ...

for i in range(nIterations):
    pass  # pass does nothing, it's a placeholder
```

## 5. Modulus Operator (%)
The modulus operator `%` returns the remainder of a division. It is often used to check if a number is even or odd.


```python
print("7 % 2 =", 7 % 2)  # Remainder is 1 (odd)
print("6 % 2 =", 6 % 2)  # Remainder is 0 (even)
```

## 6. Combining Loops and Conditionals
You can use `if-else` statements inside a loop to perform different actions based on the current value.


```python
for n in range(10):
    if n % 2 == 0:
        print(n, 'is an even number')
    else:
        print(n, 'is an odd number')
```

## 7. Practice Task: Fibonacci Series
**Exercise:**
Create a function `fibonacci(n)` that prints the first `n+3` numbers of the Fibonacci series.
- The series starts with 0, 1.
- Each subsequent number is the sum of the two preceding ones (0, 1, 1, 2, 3, 5, ...).


```python
def fibonacci(n):
    # Initialize the first two numbers
    fib_sequence = [0, 1]

    # Generate the rest of the sequence
    # We need n+3 numbers in total, so we loop n+1 times (since we already have 2)
    for i in range(n + 1):
        next_val = fib_sequence[-1] + fib_sequence[-2]
        fib_sequence.append(next_val)

    print("Fibonacci series:", fib_sequence)

# Test the function
fibonacci(5)
```

## 9. Iterating Directly Over Lists / Strings / Arrays

In Python, you can put a list (or string, or array) directly into a `for` loop — no `range()` needed. The loop variable takes each value, one at a time.

This is cleaner than using `range(len(...))`. Reach for `range()` only when you actually need a counter.

```python
# Iterating directly over a list (the cleanest way)
materials = ['steel', 'aluminum', 'titanium']
for material in materials:           # 'material' takes each value in turn
    print("Working with:", material)

# Iterating over a string — each character is one item
for char in "AE444":
    print(char)

# Iterating over a NumPy array directly
import numpy as np
voltages = np.array([3.3, 5.0, 12.0])
for v in voltages:
    print("Voltage:", v, "V")
```

## 10. The `enumerate()` Function — Index AND Value Together

Sometimes you need both the position (index) and the value at the same time. Instead of writing `for i in range(len(...))`, use `enumerate()` — it gives you both in one neat package.

```python
# enumerate() gives you (index, value) pairs
materials = ['steel', 'aluminum', 'titanium']

for i, name in enumerate(materials):
    print("Index", i, "->", name)

# You can choose the starting index (default is 0)
for i, name in enumerate(materials, start=1):
    print("Sample #", i, ":", name)

# Without enumerate, this would look uglier:
for i in range(len(materials)):
    print("Index", i, "->", materials[i])     # works, but harder to read
```

## 11. Nested For Loops

A `for` loop inside another `for` loop. The inner loop runs *completely* once for each step of the outer loop. This is useful for 2D things: tables, grids, matrices.

For an outer loop of size `m` and inner loop of size `n`, the inner code runs `m × n` times.

```python
# Multiplication table using a nested for loop
print("3x3 multiplication table:")
for i in range(1, 4):           # outer loop — rows
    for j in range(1, 4):       # inner loop — columns
        # Use end='\t' to print on the same line, separated by tabs
        print(i * j, end='\t')
    print()                     # newline after each row

# 2D grid example: print coordinates
print("\nGrid coordinates:")
for row in range(3):
    for col in range(3):
        print("(", row, ",", col, ")", end=' ')
    print()
```

## 12. Loop Accumulators — Sum, Product, Min, Max

A very common loop pattern is **accumulating** a result one item at a time. The trick: start with a sensible initial value *outside* the loop, then update it *inside* the loop.

| Goal      | Initial value | Update rule              |
|:----------|:-------------:|:-------------------------|
| Sum       | `0`           | `total += x`             |
| Product   | `1`           | `product *= x`           |
| Maximum   | `arr[0]`      | `if x > best: best = x`  |
| Counter   | `0`           | `count += 1`             |

The next cell shows several engineering examples.

```python
# Engineering example: process a list of measurements
measurements = [12.3, 9.8, 15.2, 11.0, 14.5]

# Sum accumulator
total = 0
for x in measurements:
    total += x          # add each value to the running total
print("Sum:", total)

# Product accumulator (start at 1, not 0!)
product = 1
for x in [2, 3, 4]:
    product *= x        # multiply each value into the running product
print("Product:", product)

# Find the smallest value (without min())
smallest = measurements[0]    # assume first is smallest
for x in measurements[1:]:
    if x < smallest:
        smallest = x          # update if we find a smaller one
print("Smallest:", smallest)
```

## 13. Common Pitfalls

1. **Modifying a list while iterating it** — items get skipped or you get an `IndexError`. Iterate over a copy: `for x in mylist[:]:`.

2. **Off-by-one with `range()`** — `range(1, 5)` produces `1, 2, 3, 4` (the stop is excluded). To include 5, write `range(1, 6)`.

3. **Forgetting to indent the loop body** — Python uses indentation to know what is "inside" the loop.

4. **Using `range(len(...))` when you only need values** — `for x in mylist:` is cleaner. Only use indices when you actually need them.

5. **Reusing the loop variable name as something else** — e.g., `i` for both an outer and inner loop. Always use distinct names.

```python
# Pitfall #1 demo — modifying a list while iterating over it
numbers = [1, 2, 3, 4, 5]

# DON'T DO THIS — items get skipped:
# for x in numbers:
#     if x % 2 == 0:
#         numbers.remove(x)

# Instead, iterate over a copy with [:]
for x in numbers[:]:                # the [:] makes a quick copy
    if x % 2 == 0:
        numbers.remove(x)

print("After removing evens:", numbers)

# Pitfall #2 demo — range() stop is exclusive
for i in range(1, 5):    # prints 1, 2, 3, 4 (NOT 5!)
    print(i, end=' ')
print()
```

## 14. More Practice Tasks

Try each task on paper first, then write the code.

**Task A — Factorial:** Write `factorial(n)` that returns `n! = 1·2·3·…·n`. Use a loop accumulator. (Recall: `0! = 1`.)

**Task B — Sum of Squares:** Write `sum_of_squares(n)` that returns `1² + 2² + … + n²`.

**Task C — Find Max in Array (without `max`):** Write `find_max(numbers)` that returns the largest value in a list, without using Python's built-in `max()`. Hint: assume the first item is the largest, then check the rest one by one.

```python
# Solution to Task A — Factorial
def factorial(n):
    # 0! = 1 by definition
    result = 1
    for i in range(1, n + 1):     # 1, 2, ..., n
        result *= i               # multiply running product
    return result

# Quick test
for n in [0, 1, 5, 7]:
    print(n, "! =", factorial(n))
```

```python
# Solution to Task B — Sum of Squares
def sum_of_squares(n):
    total = 0
    for i in range(1, n + 1):     # 1, 2, 3, ..., n
        total += i ** 2           # add i squared
    return total

# Test for n = 4: 1 + 4 + 9 + 16 = 30
print("Sum of squares 1..4 =", sum_of_squares(4))
print("Sum of squares 1..10 =", sum_of_squares(10))
```

```python
# Solution to Task C — Find the Maximum in an Array (without max())
def find_max(numbers):
    # Start by assuming the first item is the largest
    largest = numbers[0]
    # Compare each remaining item to 'largest'
    for x in numbers[1:]:
        if x > largest:
            largest = x   # update if we find a bigger one
    return largest

# Test with sample data
data = [3, 7, 1, 9, 4, 2]
print("Maximum =", find_max(data))
print("Verify with built-in max():", max(data))
```
