## 1. Basic For Loop
A `for` loop iterates over a sequence (like a list, tuple, or range) and executes the code block for each item.

**Example:**

```python
# Iterate 5 times (0 to 4)
for i in range(5):
    print(i)
```

## 2. Using `range()`
The `range()` function generates a sequence of numbers. It can take start, stop, and step arguments.


```python
# Iterate from 4 to 9 (10 is exclusive)
for j in range(4, 10):
    print(j, j**2)
```

## 3. Iterating Over Arrays
You can iterate over the elements of a NumPy array or list.


```python
import numpy as np

numbers = np.linspace(3, 17, 5)  # 5 numbers from 3 to 17
print("Numbers:", numbers)

for i in range(len(numbers)):
    x = numbers[i]
    print('Iteration ' + str(i+1) + ' has the value of ' + str(x))
```

## 4. Soft Coding vs. Hard Coding
- **Hard Coding:** Using fixed values directly in the code (e.g., `range(100)`). This makes the code less flexible and harder to maintain.
- **Soft Coding:** Using variables to define parameters (e.g., `range(nIterations)`). This makes the code more adaptable.

**Example:**

```python
# Soft coding example
nIterations = 5

# ... 100s of lines of code ...

for i in range(nIterations):
    pass  # pass does nothing, it's a placeholder
```

## 5. Modulus Operator (%)
The modulus operator `%` returns the remainder of a division. It is often used to check if a number is even or odd.


```python
print("7 % 2 =", 7 % 2)  # Remainder is 1 (odd)
print("6 % 2 =", 6 % 2)  # Remainder is 0 (even)
```

## 6. Combining Loops and Conditionals
You can use `if-else` statements inside a loop to perform different actions based on the current value.


```python
for n in range(10):
    if n % 2 == 0:
        print(n, 'is an even number')
    else:
        print(n, 'is an odd number')
```

## 7. Practice Task: Fibonacci Series
**Exercise:**
Create a function `fibonacci(n)` that prints the first `n+3` numbers of the Fibonacci series.
- The series starts with 0, 1.
- Each subsequent number is the sum of the two preceding ones (0, 1, 1, 2, 3, 5, ...).


```python
def fibonacci(n):
    # Initialize the first two numbers
    fib_sequence = [0, 1]

    # Generate the rest of the sequence
    # We need n+3 numbers in total, so we loop n+1 times (since we already have 2)
    for i in range(n + 1):
        next_val = fib_sequence[-1] + fib_sequence[-2]
        fib_sequence.append(next_val)

    print("Fibonacci series:", fib_sequence)

# Test the function
fibonacci(5)
```
