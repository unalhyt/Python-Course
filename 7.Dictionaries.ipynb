## Python Basics — Dictionaries

In this notebook, we will explore **dictionaries**, a powerful data structure in Python that stores data in key-value pairs. Dictionaries are similar to real-world dictionaries where you look up a word (key) to find its definition (value).

## Learning Goals
By the end of this notebook, students will be able to:
- Create dictionaries using curly braces `{}` or the `dict()` constructor.
- Understand the concept of **keys** and **values**.
- Access, add, and modify items in a dictionary.
- Use dictionary methods like `.keys()`, `.values()`, and `.items()`.
- Check for the existence of keys using the `in` operator.


## 1. Creating Dictionaries
Dictionaries are created using curly braces `{}`. Each item consists of a **key** and a **value**, separated by a colon `:`. Items are separated by commas.

**Think of it like this:** A dictionary is like a labeled filing cabinet. Each drawer has a **label** (the key) and **contents** inside (the value). You find what you need by looking at the label — not by counting drawers like you would with a list (Notebook 4).

Unlike lists, which use position numbers (indices) to access items, dictionaries use **meaningful names** (keys). This makes your code easier to read.

| Feature | List (Notebook 4) | Dictionary |
|---------|-------------------|------------|
| Created with | `[]` | `{}` |
| Access by | Index (position number) | Key (name) |
| Ordered? | Yes | Yes (Python 3.7+) |
| Duplicates? | Allowed | Keys must be unique |
| Example | `[10, 20, 30]` | `{'x': 10, 'y': 20}` |

**Example:**

````python
# Initialize an empty dictionary
D = dict()
print("Empty dictionary:", D)

# Add key-value pairs
D['name'] = 'Mike'
D['AgeRange'] = [20, 50]

print("Dictionary D:", D)
````

### Alternative Syntax
You can also create a dictionary directly by specifying key-value pairs inside curly braces.


````python
E = {'name': 'Mike', 'AgeRange': [20, 50]}
print("Dictionary E:", E)
````

## 2. Accessing Values
To retrieve a value from a dictionary, you use its key inside square brackets `[]`. This is similar to how we use an index with lists (Notebook 8), but instead of a number, we use the key name.

````python
print("Name:", D['name'])
print("Age Range:", D['AgeRange'])
````

**Safe access with `.get()`:** If you try to access a key that does not exist using `[]`, Python will raise a `KeyError`. The `.get()` method is a safer alternative — it returns `None` (or a default value you choose) instead of crashing:

````python
# Safe access with .get()
sensor = {'temperature': 85.0, 'pressure': 2.5}

# Using .get() — returns None if key doesn't exist
print("Temperature:", sensor.get('temperature'))  # 85.0
print("Humidity:", sensor.get('humidity'))          # None (no crash!)

# You can also set a default value
print("Humidity:", sensor.get('humidity', 'Not measured'))

# Compare: using [] on a missing key would crash:
# print(sensor['humidity'])  # KeyError: 'humidity'
````

## 3. Dictionary Methods
Dictionaries have several useful methods to inspect their contents.

- `.keys()`: Returns a list of all keys.
- `.values()`: Returns a list of all values.
- `.items()`: Returns a list of all key-value pairs (as tuples).


````python
print("Keys:", D.keys())
print("Values:", D.values())
print("Items:", D.items())
````

**Note:** The `.items()` method returns each key-value pair as a **tuple** (Notebook 5). For example, `('name', 'Mike')` is a tuple with two elements. You will use `.items()` heavily when looping through dictionaries in Notebook 18 (For Loops).

## 4. Checking for Keys
You can check if a key exists in a dictionary using the `in` operator. Note that this only checks keys, not values.


````python
# Check if 'name' is a key
print("'name' in D:", 'name' in D)

# Check if 'Name' (capitalized) is a key
print("'Name' in D:", 'Name' in D)

# Check if 'Mike' (a value) is a key (should be False)
print("'Mike' in D:", 'Mike' in D)
````

**Checking for values:** The `in` operator only checks keys. To check if a **value** exists, use `.values()`:

````python
# Checking for values (not keys)
print("'Mike' in D.values():", 'Mike' in D.values())  # True
print("'Jane' in D.values():", 'Jane' in D.values())  # False
````

## 5. Modifying Dictionaries
Dictionaries are **mutable** — you can change, add, and remove items after creating them.

### Updating and Adding
To update an existing value or add a new key-value pair, use square brackets `[]` with assignment `=`. If the key already exists, the value is **overwritten**. If the key is new, it is **added**.

````python
# Engineering example: updating experiment measurements
experiment = {'trial': 1, 'temperature': 22.5, 'result': 'pending'}
print("Before:", experiment)

# Update an existing value
experiment['result'] = 'success'
print("After update:", experiment)

# Add a new key-value pair
experiment['duration_sec'] = 45.2
print("After adding:", experiment)
````

### Removing Items
You can remove items using `del` or the `.pop()` method. The `.pop()` method also **returns** the removed value, which can be useful.

````python
# Removing items from a dictionary
sample = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
print("Before:", sample)

# Using del — removes the key-value pair
del sample['d']
print("After del 'd':", sample)

# Using .pop() — removes AND returns the value
removed_value = sample.pop('c')
print("Popped value:", removed_value)
print("After pop 'c':", sample)
````

## 6. Nested Dictionaries
A dictionary can contain another dictionary as a value. This is called a **nested dictionary** and is very useful for organizing complex data — like storing multiple sensor readings by location.

````python
# Nested dictionary: sensor readings by location
lab_sensors = {
    'room_101': {'temperature': 22.1, 'humidity': 45.0},
    'room_102': {'temperature': 24.8, 'humidity': 52.3},
    'room_103': {'temperature': 19.5, 'humidity': 38.7}
}

# Access the entire inner dictionary for room_101
print("Room 101 data:", lab_sensors['room_101'])

# Access a specific value: temperature in room_102
print("Room 102 temp:", lab_sensors['room_102']['temperature'])

# Add a new reading to an existing room
lab_sensors['room_101']['pressure'] = 1.01
print("Room 101 updated:", lab_sensors['room_101'])
````

## 7. Common Pitfalls

**1. Using mutable types as keys**

Dictionary keys must be **immutable** (cannot be changed). Strings, numbers, and tuples (Notebook 5) work as keys. Lists (Notebook 4) do **NOT** work because they are mutable.

```python
# This works:
d = {(1, 2): 'point'}     # tuple key — OK

# This crashes:
# d = {[1, 2]: 'point'}   # TypeError: unhashable type: 'list'
```

**2. Accessing a key that doesn't exist**

Using `[]` on a missing key causes a `KeyError`. Use `.get()` instead (see Section 2) or check with `in` first.

**3. Keys are case-sensitive**

`'Name'` and `'name'` are **different keys**. This is a very common source of bugs. Be consistent with capitalization.

**4. Accidentally overwriting a key**

If you assign a value to a key that already exists, the old value is **silently replaced** — Python will not warn you. Double-check your key names when adding data.

````python
# Pitfall 3: case-sensitive keys
data = {'Temperature': 100}
# data['temperature']  # KeyError! Capital T vs lowercase t

# Pitfall 4: accidentally overwriting
scores = {'math': 90, 'science': 85}
scores['math'] = 75  # Silently replaces 90 with 75!
print("Scores after overwrite:", scores)
````
