## Python Basics — Dictionaries

In this notebook, we will explore **dictionaries**, a powerful data structure in Python that stores data in key-value pairs. Dictionaries are similar to real-world dictionaries where you look up a word (key) to find its definition (value).

## Learning Goals
By the end of this notebook, students will be able to:
- Create dictionaries using curly braces `{}` or the `dict()` constructor.
- Understand the concept of **keys** and **values**.
- Access, add, and modify items in a dictionary.
- Use dictionary methods like `.keys()`, `.values()`, and `.items()`.
- Check for the existence of keys using the `in` operator.


## 1. Creating Dictionaries
Dictionaries are created using curly braces `{}`. Each item consists of a key and a value, separated by a colon `:`. Items are separated by commas.

**Example:**

```python
# Initialize an empty dictionary
D = dict()
print("Empty dictionary:", D)

# Add key-value pairs
D['name'] = 'Mike'
D['AgeRange'] = [20, 50]

print("Dictionary D:", D)
```

### Alternative Syntax
You can also create a dictionary directly by specifying key-value pairs inside curly braces.


```python
E = {'name': 'Mike', 'AgeRange': [20, 50]}
print("Dictionary E:", E)
```

## 2. Accessing Values
To retrieve a value from a dictionary, you use its key inside square brackets `[]`.


```python
print("Name:", D['name'])
print("Age Range:", D['AgeRange'])
```

## 3. Dictionary Methods
Dictionaries have several useful methods to inspect their contents.

- `.keys()`: Returns a list of all keys.
- `.values()`: Returns a list of all values.
- `.items()`: Returns a list of all key-value pairs (as tuples).


```python
print("Keys:", D.keys())
print("Values:", D.values())
print("Items:", D.items())
```

## 4. Checking for Keys
You can check if a key exists in a dictionary using the `in` operator. Note that this only checks keys, not values.


```python
# Check if 'name' is a key
print("'name' in D:", 'name' in D)

# Check if 'Name' (capitalized) is a key
print("'Name' in D:", 'Name' in D)

# Check if 'Mike' (a value) is a key (should be False)
print("'Mike' in D:", 'Mike' in D)
```

## 5. Practice Task
**Exercise:**
1.  Ask the user for two numbers.
2.  Create a dictionary with keys `'firstNum'`, `'secondNum'`, and `'product'`.
3.  Store the user's numbers and their product in the dictionary.
4.  Print the dictionary items.


```python
# Clear the previous dictionary variable (optional, for demonstration)
if 'D' in locals():
    del D

# Initialize a new dictionary
D = dict()

# Get input from the user
num1 = float(input('Enter first number: '))
D['firstNum'] = num1

num2 = float(input('Enter second number: '))
D['secondNum'] = num2

# Calculate product
D['product'] = num1 * num2

# Print the result
print("Dictionary items:", D.items())
```
