## Python Basics — While Loops

In this notebook, we will explore **while loops**, which repeat a block of code as long as a specified condition is true. While loops are useful when you don't know in advance how many times you need to iterate.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `while` loops to repeat code based on a condition.
- Understand the difference between `for` loops (known iterations) and `while` loops (unknown iterations).
- Use `break` to exit a `while` loop.
- Implement a Poisson-like counter using a `while` loop.


## 1. Basic While Loop
A `while` loop continues to execute as long as the condition evaluates to `True`.

**Example:**

````python
toggle = True
i = 0

while toggle:
    print(i)
    i += 1
    if i == 8:
        toggle = False  # Condition becomes False, loop stops
````

## 2. Infinite Loops & Break
Be careful with `while` loops! If the condition never becomes `False`, the loop will run forever (infinite loop). You can use `break` to exit manually.


````python
i = 0
while True:  # Infinite loop
    print(i)
    i += 1
    if i == 8:
        break  # Exit the loop
````

## 3. While Loop with Condition
You can also put the condition directly in the `while` statement.


````python
i = 0
while i < 8:
    print(i)
    i += 1
````

## 4. For Loop vs. While Loop
- **For Loop:** Use when you know the number of iterations in advance (e.g., iterating over a list).
- **While Loop:** Use when the number of iterations depends on a condition (e.g., waiting for user input, optimization algorithms).


## 5. Practice Task: Poisson Counter
**Exercise:**
Create a function `poissonCounter(lam)` that simulates a Poisson process.
- Start with a value of 1.
- Multiply the value by a random number between 0 and 1 (uniform distribution) repeatedly.
- Stop when the value is less than `exp(-lam)`.
- Return the number of multiplications performed.


````python
import numpy as np

def poissonCounter(lam):
    # Initialize variables
    counter = 0
    currval = 1
    target = np.exp(-lam)

    # Run the algorithm
    while currval >= target:
        currval *= np.random.uniform(0, 1)
        counter += 1

    return counter

# Test the function
print("Counter result for lambda=10:", poissonCounter(10))

# Run multiple simulations
p = np.zeros(100)
for i in range(100):
    p[i] = poissonCounter(10)

print("Results of 100 simulations:", p)
````



## 8. Counter Patterns — Increment and Decrement

The most basic `while` pattern is a **counter**: a variable that goes up (or down) by a fixed step on each iteration. Three common shortcuts:

| Shortcut | Equivalent to | Meaning           |
|:--------:|:--------------|:------------------|
| `i += 1` | `i = i + 1`   | increment by 1    |
| `i -= 1` | `i = i - 1`   | decrement by 1    |
| `i += 2` | `i = i + 2`   | step up by 2      |
| `i *= 2` | `i = i * 2`   | double each step  |

The example below shows each pattern.

````python
# Increment counter — count up from 0
i = 0
while i < 5:
    print("Count up:", i)
    i += 1                  # same as i = i + 1

# Decrement counter — count down to 0
i = 5
while i > 0:
    print("Count down:", i)
    i -= 1                  # same as i = i - 1

# Step by 2 — even numbers from 0 to 10
i = 0
while i <= 10:
    print("Even:", i)
    i += 2
````

## 9. While Loops with Sentinel Values

A **sentinel** is a special value that signals "stop". For example, in a program that asks the user to enter numbers, the sentinel might be `'quit'`.

The idea:
```python
while user_input != 'quit':
    user_input = input("Enter a number (or 'quit'): ")
    ...
```

Below we simulate this pattern without `input()` (which can hang in non-interactive notebooks).

````python
# Sentinel pattern using a fixed list of "user inputs" (input() commented out)
# In a real program you would do:
#   user_input = input("Enter a number (or 'quit' to stop): ")

simulated_inputs = ["3", "5", "abc", "7", "quit"]   # pretend user types these
total = 0

for user_input in simulated_inputs:
    if user_input == "quit":            # the sentinel value
        break
    # try to convert to a number
    if user_input.replace('.', '', 1).lstrip('-').isdigit():
        total += float(user_input)
        print("Added", user_input, "-> running total:", total)
    else:
        print("Ignoring invalid input:", user_input)

print("Final total:", total)
````

## 10. Convergence Loops — A Classic Engineering Pattern

Engineers often use `while` loops in iterative algorithms (Newton's method, fixed-point iteration, optimization). The idea: keep iterating until the error is small enough.

```
while error > tolerance:
    update x
    compute new error
```

Always include a **safety counter** (max iterations) to avoid an infinite loop if the algorithm fails to converge.

````python
# Engineering example: solve cos(x) = x (a classic fixed-point iteration)
import numpy as np

x = 0.5             # initial guess
tol = 1e-8          # how close is "close enough"
max_iters = 100     # safety: never run more than 100 iterations
iters = 0
error = 1.0         # any value > tol so we enter the loop

while error > tol and iters < max_iters:
    x_new = np.cos(x)         # the iteration step
    error = abs(x_new - x)    # how much did x change?
    x = x_new
    iters += 1

print("Solution: x =", x, "after", iters, "iterations")
print("Check: cos(x) =", np.cos(x))
````

## 11. Do-While Equivalent in Python

Some languages (C, Java) have a "do-while" loop that always runs the body **at least once** before checking the condition. Python does not have this, but you can simulate it with `while True:` + `break`:

```python
while True:
    do_something()
    if exit_condition:
        break
```

This pattern is also useful when the test naturally happens *after* an action — like rolling a die until you roll a 6.

````python
# Do-while pattern in Python — always execute the body at least once
import numpy as np

# Roll a die until we get a 6 (we always roll at least once)
np.random.seed(42)
rolls = 0
while True:                       # the "do" part: enter unconditionally
    roll = np.random.randint(1, 7)   # 1 through 6
    rolls += 1
    print("Rolled", roll)
    if roll == 6:                 # the "while" condition tested at the bottom
        break

print("Total rolls until a 6:", rolls)
````

## 12. Common Pitfalls

1. **Forgetting to update the counter** — `while i < 10:` with no `i += 1` inside means the loop never ends. Always trace through one iteration in your head.

2. **Float comparison** — `while x != 0.1:` can loop forever because of tiny floating-point errors. Use a tolerance: `while abs(x - 0.1) > 1e-9:`.

3. **Off-by-one** — Be careful whether the loop should run while `i < n` or `i <= n`. They differ by one iteration!

4. **Always add a safety counter for convergence loops** — In real engineering code, also include a max-iteration check so a non-converging algorithm doesn't run forever:
   ```python
   while error > tol and iters < max_iters:
       ...
   ```

````python
# Pitfall #1 — forgetting to update the counter (DO NOT RUN as-is, would loop forever)
# i = 0
# while i < 5:
#     print(i)        # i is never incremented => infinite loop!

# Safe version:
i = 0
while i < 5:
    print("i =", i)
    i += 1            # always remember to update i!

# Pitfall #2 — float comparisons can be tricky
# Bad: while x != 0.1 (might never become exactly 0.1)
x = 0.0
count = 0
while abs(x - 0.3) > 1e-9 and count < 100:    # use tolerance + safety counter
    x += 0.1
    count += 1
print("Reached", x, "in", count, "steps")
````

## 13. More Practice Tasks

**Task A — Sum of Digits:** Write `sum_of_digits(n)` that returns the sum of all digits of an integer `n`. Use a `while` loop and the `%` and `//` operators. Example: `sum_of_digits(1234) == 10`.

**Task B — Guess-the-Number Simulator:** A "secret" number between 1 and 100 is set. Write a function that simulates a player guessing the midpoint of the remaining range each round, narrowing it after each guess until correct. Return the number of attempts.

**Task C — Square Root via Newton's Method:** Write `my_sqrt(n)` that approximates `√n` using the iteration `x = 0.5 * (x + n/x)`. Stop when `|x_new - x| < 1e-10`. Compare to `math.sqrt(n)`.

````python
# Solution to Task A — Sum of Digits
def sum_of_digits(n):
    n = abs(n)              # handle negatives
    total = 0
    while n > 0:
        total += n % 10     # last digit
        n = n // 10         # remove the last digit
    return total

# Tests
for value in [1234, 9, 1000, -42]:
    print("sum_of_digits(", value, ") =", sum_of_digits(value))
````

````python
import numpy as np

# Solution to Task B — Guess-the-Number Simulator
def guess_simulator(secret, low=1, high=100):
    # The "player" guesses the midpoint each round (binary search style)
    attempts = 0
    while True:
        attempts += 1
        guess = (low + high) // 2
        if guess == secret:
            print("Got it in", attempts, "attempts!")
            return attempts
        elif guess < secret:
            low = guess + 1     # narrow range upward
        else:
            high = guess - 1    # narrow range downward

# Test
np.random.seed(0)
guess_simulator(secret=42)
guess_simulator(secret=99)
````

````python
# Solution to Task C — Square Root via Newton's Method
def my_sqrt(n, tol=1e-10):
    # Newton's method: x_{k+1} = 0.5 * (x_k + n/x_k)
    x = n / 2.0           # an initial guess
    error = 1.0           # any value bigger than tol to enter the loop
    iters = 0
    while error > tol:
        x_new = 0.5 * (x + n / x)
        error = abs(x_new - x)    # how much did x change?
        x = x_new
        iters += 1
    return x, iters

# Test against math.sqrt
import math
for value in [2, 25, 1000]:
    root, k = my_sqrt(value)
    print("sqrt(", value, ") =", root, "  (in", k, "iters)  vs math.sqrt:", math.sqrt(value))
````
