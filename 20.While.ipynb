## Python Basics — While Loops

In this notebook, we will explore **while loops**, which repeat a block of code as long as a specified condition is true. While loops are useful when you don't know in advance how many times you need to iterate.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `while` loops to repeat code based on a condition.
- Understand the difference between `for` loops (known iterations) and `while` loops (unknown iterations).
- Use `break` to exit a `while` loop.
- Implement a Poisson-like counter using a `while` loop.


## 1. Basic While Loop
A `while` loop continues to execute as long as the condition evaluates to `True`.

**Example:**

````python
toggle = True
i = 0

while toggle:
    print(i)
    i += 1
    if i == 8:
        toggle = False  # Condition becomes False, loop stops
````

## 2. Infinite Loops & Break
Be careful with `while` loops! If the condition never becomes `False`, the loop will run forever (infinite loop). You can use `break` to exit manually.


````python
i = 0
while True:  # Infinite loop
    print(i)
    i += 1
    if i == 8:
        break  # Exit the loop
````

## 3. While Loop with Condition
You can also put the condition directly in the `while` statement.


````python
i = 0
while i < 8:
    print(i)
    i += 1
````

## 4. For Loop vs. While Loop
- **For Loop:** Use when you know the number of iterations in advance (e.g., iterating over a list).
- **While Loop:** Use when the number of iterations depends on a condition (e.g., waiting for user input, optimization algorithms).


## 5. Practice Task: Poisson Counter
**Exercise:**
Create a function `poissonCounter(lam)` that simulates a Poisson process.
- Start with a value of 1.
- Multiply the value by a random number between 0 and 1 (uniform distribution) repeatedly.
- Stop when the value is less than `exp(-lam)`.
- Return the number of multiplications performed.


````python
import numpy as np

def poissonCounter(lam):
    # Initialize variables
    counter = 0
    currval = 1
    target = np.exp(-lam)

    # Run the algorithm
    while currval >= target:
        currval *= np.random.uniform(0, 1)
        counter += 1

    return counter

# Test the function
print("Counter result for lambda=10:", poissonCounter(10))

# Run multiple simulations
p = np.zeros(100)
for i in range(100):
    p[i] = poissonCounter(10)

print("Results of 100 simulations:", p)
````

## 6. Code Readability
While you can write concise code, readability is often more important. Compare the two versions of `poissonCounter`:

**Readable Version:**
```python
def poissonCounter(lam):
    counter = 0
    currval = 1
    target = np.exp(-lam)
    while currval >= target:
        currval *= np.random.uniform(0, 1)
        counter += 1
    return counter
```

**Concise (but harder to read) Version:**
```python
def poissonCounter(l):
    k, p, t = 0, 1, np.exp(-l)
    while p > t:
        p *= np.random.uniform(0, 1)
        k += 1
    return k
```
