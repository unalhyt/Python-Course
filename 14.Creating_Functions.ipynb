## 1. Defining a Function
A function is defined using the `def` keyword, followed by the function name and parentheses `()`. The code block inside the function is indented.

**Example:**

```python
def awesomeFunction():
    print(1 + 1)

# Calling the function
awesomeFunction()
```

**Naming Convention:** In Python, function names should use `snake_case` (lowercase words separated by underscores). This makes them easy to read. For example: `calculate_area`, `convert_temperature`, `get_average`. Avoid names like `awesomeFunction` (that is camelCase, used in other languages like Java).

```python
# A function with a descriptive snake_case name
def calculate_circle_area(radius):
    """Calculate the area of a circle given its radius."""
    area = 3.14159 * radius ** 2
    return area

# Call the function
print("Area of circle with radius 5:", calculate_circle_area(5))
```

## 2. Function Arguments
Functions can accept inputs, called arguments. These are specified inside the parentheses.


```python
def superawesomeFunction(input1, input2):
    return input1 + input2

# Calling the function with arguments
result = superawesomeFunction(12, 2)
print("Result:", result)
```

```python
# What happens if you pass the wrong number of arguments?
try:
    result = superawesomeFunction(12)  # Missing second argument
except TypeError as e:
    print("Error:", e)
    print("Fix: Make sure you pass all required arguments.")
```

## 3. Return Values
Functions can return a value using the `return` statement. If no return statement is provided, the function returns `None`.


```python
output = awesomeFunction()
print("Output of awesomeFunction:", output)  # None

output = superawesomeFunction(12, 2)
print("Output of superawesomeFunction:", output)  # 14
```

**Key difference: `print()` vs `return`**

Think of it this way:
- `print()` is like **saying** the answer out loud — everyone can hear it, but nobody can use it later
- `return` is like **handing** the answer to someone — they can store it and use it in further calculations

If you forget `return`, your function will return `None` by default. Also note that `return` **immediately exits** the function — any code after `return` will not run.

```python
# Demonstrating that return exits the function immediately
def test_return():
    print("This line runs")    # This will print
    return 42
    print("This line is NEVER reached")  # This will NOT print

result = test_return()
print("Returned value:", result)
```

## 4. Multiple Outputs
Functions can return multiple values as a tuple.


```python
def funfun(in1, in2):
    sumres = in1 + in2
    prodres = in1 * in2
    print('Their sum is ' + str(sumres))
    print('Their product is ' + str(prodres))
    return sumres, prodres

# Unpacking the returned tuple
outS, outP = funfun(4, 5)
print("Sum:", outS)
print("Product:", outP)
```

## 4.5 Default Arguments

Sometimes you want a function argument to have a **default value** — a value that is used automatically if the caller does not provide one. This makes the function more flexible and easier to use.

Think of it like ordering coffee: if you do not specify a size, you get the default (medium). But you can always override it.

```python
# Function with a default argument
def convert_temperature(celsius, to_unit="fahrenheit"):
    """Convert Celsius to Fahrenheit or Kelvin."""
    if to_unit == "fahrenheit":
        return celsius * 9/5 + 32
    elif to_unit == "kelvin":
        return celsius + 273.15

# Using the default (Fahrenheit)
print("100°C in Fahrenheit:", convert_temperature(100))

# Overriding the default
print("100°C in Kelvin:", convert_temperature(100, "kelvin"))
```

## 4.6 Keyword Arguments

When calling a function, you can specify arguments by **name** (keyword) instead of just by position. This makes your code easier to read, especially when a function has many arguments.

```python
def describe_experiment(name, temperature, pressure, duration):
    """Print a summary of an experiment."""
    print(f"Experiment: {name}")
    print(f"  Temperature: {temperature}°C")
    print(f"  Pressure: {pressure} atm")
    print(f"  Duration: {duration} hours")

# Positional arguments — order matters
describe_experiment("Heat Test", 150, 2.5, 3)

print()  # Blank line for readability

# Keyword arguments — order does not matter
describe_experiment(
    duration=3,
    name="Heat Test",
    pressure=2.5,
    temperature=150
)
```

## 4.7 Docstrings — Documenting Your Functions

A **docstring** is a string written as the very first line inside a function. It describes what the function does. You can then read it using `help()` (as we learned in Notebook 13).

Docstrings are written between triple quotes `"""..."""`.

```python
def compute_velocity(distance, time):
    """
    Compute velocity from distance and time.

    Arguments:
        distance -- distance traveled (in meters)
        time -- time elapsed (in seconds)

    Returns:
        velocity in meters per second (m/s)
    """
    return distance / time

# Now help() shows our docstring!
help(compute_velocity)

# Use the function
print("Velocity:", compute_velocity(100, 9.58), "m/s")
```

## 5. Lambda Functions
Lambda functions are small, anonymous functions defined with the `lambda` keyword. They are often used for short, simple operations.


```python
# Define a lambda function: f(x) = x^2 - 1
littlefun = lambda x: x**2 - 1

# Call the lambda function
print("Result of littlefun(4):", littlefun(4))
```

```python
# Practical use of lambda: sorting a list of tuples
# Each tuple is (experiment_name, temperature)
experiments = [("Test A", 85), ("Test B", 42), ("Test C", 120)]

# Sort by temperature (the second element) using lambda
sorted_experiments = sorted(experiments, key=lambda x: x[1])
print("Sorted by temperature:")
for name, temp in sorted_experiments:
    print(f"  {name}: {temp}°C")
```

## 6. Practice Task: Factorial Function
**Exercise:**
1.  Create a function `myfactorial(val)` that computes the factorial of an integer `val`.
2.  Compare your result with the `math.factorial()` function from the `math` module.

Hint: $n! = n \times (n-1) \times \dots \times 1$.


```python
import numpy as np
import math

# Define the factorial function
def myfactorial(val):
    # Create an array from 1 to val
    numbers = np.arange(1, val + 1)
    # Calculate the product of all numbers
    return np.prod(numbers)

# Test the function
val = 4
print("My factorial of 4:", myfactorial(val))
print("Math factorial of 4:", math.factorial(val))

# Test with a larger number
k = 23
print(f"My factorial of {k}: {myfactorial(k)}")
print(f"Math factorial of {k}: {math.factorial(k)}")

# Note: For very large numbers, floating-point precision issues might occur with np.prod.
# math.factorial uses arbitrary-precision integers, which is safer for factorials.
```

## 7. Common Pitfalls

1. **Forgetting `return`:**
   - If your function calculates a value but uses `print()` instead of `return`, you cannot store or reuse the result. The function will return `None`.
   - Example: `result = my_function()` will be `None` if `my_function` does not have a `return` statement.

2. **Confusing `print()` with `return`:**
   - `print()` displays text on the screen. `return` sends a value back to the caller. They are not the same!
   - A common beginner mistake is to `print()` inside the function and then wonder why the result is `None`.

3. **Wrong argument order:**
   - If you use positional arguments, the order must match the function definition. Use keyword arguments to avoid this.
   - Example: `convert_temperature("kelvin", 100)` would fail — the arguments are in the wrong order.

4. **Modifying a list inside a function:**
   - Lists are passed by reference. If you modify a list inside a function, the original list is also changed! This can cause unexpected bugs.
   
5. **Using the same name for a function and a variable:**
   - If you define `sum = 10`, you can no longer use the built-in `sum()` function in that scope. Be careful with naming!

## 9. Task Solutions

```python
# Task 1 Solution
def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return celsius * 9/5 + 32

# Test
print("0°C =", celsius_to_fahrenheit(0), "°F")    # 32.0
print("100°C =", celsius_to_fahrenheit(100), "°F")  # 212.0
```

```python
# Task 2 Solution
def compute_statistics(data):
    """Return the min, max, and average of a list of numbers."""
    minimum = min(data)
    maximum = max(data)
    average = sum(data) / len(data)
    return minimum, maximum, average

# Test
sensor_data = [23.1, 45.7, 12.3, 67.8, 34.2]
lo, hi, avg = compute_statistics(sensor_data)
print(f"Min: {lo}, Max: {hi}, Average: {avg:.2f}")
```

```python
# Task 3 Solution
def is_even(number):
    """Return True if number is even, False if odd."""
    return number % 2 == 0

# Test with numbers 1 to 10
for n in range(1, 11):
    if is_even(n):
        print(f"{n} is even")
```

```python
# Task 4 Solution
def compute_drag_force(velocity, drag_coeff=0.47, air_density=1.225, area=0.0045):
    """
    Calculate aerodynamic drag force.
    Fd = 0.5 * rho * v^2 * Cd * A
    """
    return 0.5 * air_density * velocity**2 * drag_coeff * area

# Test with just velocity (uses defaults for a sphere in air)
print("Drag at 10 m/s:", round(compute_drag_force(10), 4), "N")
print("Drag at 50 m/s:", round(compute_drag_force(50), 4), "N")

# Override defaults for a different object
print("Drag at 50 m/s (large area):",
      round(compute_drag_force(50, area=0.5), 4), "N")
```
