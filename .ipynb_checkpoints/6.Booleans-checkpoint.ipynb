## Python Basics — Booleans

In this notebook, we will explore **Booleans**, a fundamental data type in Python that represents truth values: `True` or `False`. Booleans are essential for making decisions in your code.

## Learning Goals
By the end of this notebook, students will be able to:
- Understand the concept of Boolean values (`True` and `False`).
- Use comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) to generate Booleans.
- Combine multiple conditions using logical operators (`and`, `or`, `not`).
- Convert other data types to Booleans using `bool()`.
- Apply Boolean logic to solve simple problems.


## 1. Boolean Values
A Boolean value can only be `True` or `False`. Note that the first letter **must** be capitalized.

**Think of it like this:** A Boolean is like a light switch — it can only be ON (`True`) or OFF (`False`). There is no in-between. In engineering, Booleans are like binary signals: 1 or 0, yes or no, pass or fail.

**Example:**

````python
boolTrue = True
strTrue = 'True'  # This is a string, not a boolean

print("Boolean True:", boolTrue)
print("String 'True':", strTrue)

# Check types
print("Type of boolTrue:", type(boolTrue))
print("Type of strTrue:", type(strTrue))
````

**Careful with capitalization!** Writing `true` or `false` (lowercase) will cause a `NameError`:

````python
# Correct: capital T and F
print("True:", True)
print("False:", False)

# Incorrect (uncomment to see the error):
# print(true)   # NameError: name 'true' is not defined
````

**Fun fact:** In Python, `True` and `False` are actually special integers! `True` equals `1` and `False` equals `0`. This means you can do math with them:

````python
# Booleans are secretly integers!
print("True + True:", True + True)       # 1 + 1 = 2
print("True * 5:", True * 5)             # 1 * 5 = 5
print("False + 10:", False + 10)         # 0 + 10 = 10
print("True == 1:", True == 1)           # True
print("False == 0:", False == 0)         # True
````

## 2. Comparison Operators
Comparison operators allow you to compare two values. The result of a comparison is always a Boolean.

- `==`: Equal to
- `!=`: Not equal to
- `>`: Greater than
- `<`: Less than
- `>=`: Greater than or equal to
- `<=`: Less than or equal to


````python
# Is 4 equal to 4?
print("4 == 4:", 4 == 4)

# Is 4 greater than or equal to 5?
print("4 >= 5:", 4 >= 5)

# Is 4 NOT equal to 5?
print("4 != 5:", 4 != 5)

# Using variables
a = 8
b = 5
outcome = a == b * 2  # Is 8 equal to 10?
print("Outcome:", outcome)
````

Here is a summary table of all comparison operators:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `5 <= 3` | `False` |

**Engineering example:** Imagine you are checking sensor data from an experiment. You want to know if a temperature reading exceeds a safety limit:

````python
# Engineering example: checking sensor data
max_temperature = 150.0       # safety limit in degrees Celsius
sensor_reading = 163.5        # current reading from the sensor

# Is the reading above the limit?
is_overheating = sensor_reading > max_temperature
print("Sensor reading:", sensor_reading, "C")
print("Safety limit:", max_temperature, "C")
print("Is overheating?", is_overheating)
````

## 3. Logical Operators
You can combine multiple conditions using `and`, `or`, and `not`.

- `and`: Returns `True` if **both** conditions are true.
- `or`: Returns `True` if **at least one** condition is true.
- `not`: Reverses the boolean value.


````python
# Using 'and'
print("14 > 5 and 7 < 10 and 3 == 3:", 14 > 5 and 7 < 10 and 3 == 3)

# Using 'or'
print("14 > 5 or 7 > 10:", 14 > 5 or 7 > 10)
````

````python
# Using 'not' — reverses the boolean value
print("not True:", not True)
print("not False:", not False)
print("not (4 > 5):", not (4 > 5))   # 4 > 5 is False, so not False = True
````

Here are the **truth tables** for `and`, `or`, and `not`. These show every possible combination:

| A | B | `A and B` | `A or B` | `not A` |
|---|---|-----------|----------|---------|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

**Engineering example:** In a chemical plant, you might need to check that both temperature AND pressure are within safe limits before allowing a reaction to proceed:

````python
# Engineering example: checking multiple safety conditions
temperature = 85.0   # degrees Celsius
pressure = 2.5       # atmospheres

# Define safe ranges
temp_ok = temperature < 100.0    # temperature must be below 100 C
pressure_ok = pressure < 3.0     # pressure must be below 3 atm

# Both conditions must be True for the system to be safe
system_safe = temp_ok and pressure_ok
print("Temperature OK?", temp_ok)
print("Pressure OK?", pressure_ok)
print("System safe?", system_safe)
````

## 4. Converting to Booleans
You can convert other data types to Booleans using the `bool()` function.

- Numbers: `0` converts to `False`, any other number converts to `True`.
- Strings: An empty string `""` converts to `False`, any other string converts to `True`.


````python
print("bool(0):", bool(0))
print("bool(1):", bool(1))
print("bool(123.45):", bool(123.45))

print("bool('asd'):", bool('asd'))
print("bool(' '):", bool(' '))  # Space is not empty
print("bool(''):", bool(''))    # Empty string
````

**Truthy vs Falsy:** Python calls values that convert to `True` **"truthy"** and values that convert to `False` **"falsy"**. This also works with containers like lists (Notebook 4) and tuples (Notebook 5) — empty containers are falsy, non-empty ones are truthy:

````python
# Truthy and falsy with containers
print("bool([1, 2, 3]):", bool([1, 2, 3]))  # Non-empty list = True
print("bool([]):", bool([]))                  # Empty list = False
print("bool((1, 2)):", bool((1, 2)))          # Non-empty tuple = True
print("bool(()):", bool(()))                  # Empty tuple = False
print("bool(None):", bool(None))              # None is always False
````

## 5. Common Pitfalls

Here are some common mistakes beginners make with Booleans:

**1. Using `=` instead of `==`**

Remember: `=` is for **assigning** a value to a variable (we learned this in Notebook 1). `==` is for **comparing** two values.

**2. Confusing `and`/`or` with `&`/`|`**

In Python, use the words `and`, `or`, `not` for Boolean logic. The symbols `&`, `|` are **bitwise operators** — they work on individual bits of numbers and behave differently. Stick with `and`, `or`, `not` for now.

**3. Comparing floating-point numbers with `==`**

Due to how computers store decimal numbers, floating-point math can produce tiny rounding errors. This means `==` might give unexpected results:

**4. Writing `True`/`False` with wrong capitalization**

`true`, `false`, `TRUE`, `FALSE` are NOT valid Booleans in Python. Only `True` and `False` (with a capital first letter) work.

````python
# Pitfall 1: = vs ==
x = 5          # This ASSIGNS the value 5 to x
print("x == 5:", x == 5)   # This COMPARES x to 5 and returns True

# Pitfall 3: floating-point comparison surprise
result = 0.1 + 0.2          # You would expect 0.3
print("0.1 + 0.2 =", result)
print("0.1 + 0.2 == 0.3:", result == 0.3)  # False! Tiny rounding error

# A safer way: check if the difference is very small
print("Close enough?", abs(result - 0.3) < 0.0001)  # True
````

### Task 2: Leap Year Checker
A year is a leap year if:
- It is divisible by 4, **AND**
- It is NOT divisible by 100, **OR** it is divisible by 400.

Given the variable `year = 2024`, use comparison and logical operators to create a Boolean variable `is_leap` that is `True` if the year is a leap year. Print the result.

Test your logic with these years: 2024 (leap), 1900 (not leap), 2000 (leap).

````python
# Task 2 Solution: Leap Year Checker
year = 2024  # Change this to test other years

# Check leap year rules using logical operators
is_leap = (year % 4 == 0) and (year % 100 != 0 or year % 400 == 0)
print("Year:", year)
print("Is leap year?", is_leap)
````

### Task 3: Temperature Safety Checker
A laboratory oven is considered safe when the temperature is between 50 and 200 degrees Celsius (inclusive).

Given these three sensor readings, create a Boolean variable for each one that is `True` if the reading is in the safe range, and `False` otherwise. Use comparison operators and `and`.

```python
reading_1 = 75.0
reading_2 = 210.5
reading_3 = 50.0
```

Print each reading and whether it is safe.

````python
# Task 3 Solution: Temperature Safety Checker
reading_1 = 75.0    # sensor 1
reading_2 = 210.5   # sensor 2
reading_3 = 50.0    # sensor 3

# Check if each reading is in the safe range (50 to 200 inclusive)
safe_1 = reading_1 >= 50 and reading_1 <= 200
safe_2 = reading_2 >= 50 and reading_2 <= 200
safe_3 = reading_3 >= 50 and reading_3 <= 200

print("Reading 1:", reading_1, "C -> Safe?", safe_1)
print("Reading 2:", reading_2, "C -> Safe?", safe_2)
print("Reading 3:", reading_3, "C -> Safe?", safe_3)
````

### Task 4: Student Grade Eligibility (Challenging)
A student is eligible for honors if ALL of the following are true:
- Their exam score is at least 85
- Their lab score is at least 70
- They have NOT missed more than 2 classes

Given these variables, determine if the student is eligible:

```python
exam_score = 92
lab_score = 78
missed_classes = 1
```

Create a single Boolean variable `is_eligible` that combines all three conditions. Print the result.

````python
# Task 4 Solution: Student Grade Eligibility
exam_score = 92       # out of 100
lab_score = 78        # out of 100
missed_classes = 1    # number of missed classes

# Check all three conditions with 'and'
is_eligible = exam_score >= 85 and lab_score >= 70 and missed_classes <= 2

print("Exam score:", exam_score)
print("Lab score:", lab_score)
print("Missed classes:", missed_classes)
print("Eligible for honors?", is_eligible)
````

## 7. Mini-Quiz

Test your understanding! Try to answer these without running the code first.

**Q1:** What is the result of `not (5 > 3 and 2 > 8)`?

**Q2:** What does `bool("")` return?

**Q3:** What is the result of `True + True + False`?

**Q4:** What is the result of `10 != 10 or 3 <= 3`?

````python
# Mini-Quiz Answers — run this cell to check!

# Q1: not (5 > 3 and 2 > 8)
# Step by step: 5 > 3 = True, 2 > 8 = False
# True and False = False, not False = True
print("Q1:", not (5 > 3 and 2 > 8))    # True

# Q2: bool("") — empty string is falsy
print("Q2:", bool(""))                   # False

# Q3: True + True + False = 1 + 1 + 0
print("Q3:", True + True + False)        # 2

# Q4: 10 != 10 = False, 3 <= 3 = True
# False or True = True
print("Q4:", 10 != 10 or 3 <= 3)        # True
````
