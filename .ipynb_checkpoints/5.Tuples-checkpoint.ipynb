## Python Basics — Tuples

In this notebook, we will explore **tuples**, another fundamental data structure in Python. Tuples are similar to lists but have one key difference: they are **immutable**.

## Learning Goals
By the end of this notebook, students will be able to:
- Create tuples using parentheses `()`.
- Understand the difference between **mutable** (lists) and **immutable** (tuples) data structures.
- Access elements in a tuple.
- Convert between lists and tuples.
- Recognize when to use tuples instead of lists.


## 1. Creating Tuples
Tuples are created using parentheses `()`. Elements inside a tuple are separated by commas.

**Example:**

```python
# A tuple of mixed data types
atuple = (3, '4', 3)
print("Tuple:", atuple)

# A list with the same elements
alist = [3, '4', 3]
print("List:", alist)
```

## 2. Tuple Operations
Tuples support some operations similar to lists:
- `*` repeats a tuple.
- `.count(item)` returns the number of times an item appears.
- `.index(item)` returns the index of the first occurrence of an item.


```python
# Repeat the tuple 3 times
print("Repeated tuple:", atuple * 3)

# Count occurrences of 3
print("Count of 3:", atuple.count(3))
```

## 3. Immutability: The Key Difference
The most important difference between lists and tuples is that **tuples are immutable**. This means you cannot change, add, or remove elements after the tuple is created.

**Example:**

```python
print("Original list:", alist)
alist[0] = 5  # This is allowed for lists
print("Modified list:", alist)

# Uncommenting the line below will raise a TypeError because tuples are immutable:
# atuple[0] = 5

# However, you can reassign the entire variable to a new tuple:
atuple = (5, '4', 3)
print("New tuple assigned to variable:", atuple)
```

## 4. Converting Between Lists and Tuples
You can convert a tuple to a list using `list()` and a list to a tuple using `tuple()`. This is useful if you need to modify a tuple temporarily.


```python
print("Original tuple type:", type(atuple))

# Convert tuple to list
v = list(atuple)
print("Converted to list:", v)
print("Type of converted variable:", type(v))

# Now we can modify the list
v[0] = 254456
print("Modified list:", v)

# Convert back to tuple if needed
new_tuple = tuple(v)
print("Converted back to tuple:", new_tuple)
```

## 5. Why Use Tuples?
- **Data Integrity:** Use tuples when you want to ensure that the data cannot be changed accidentally.
- **Performance:** Tuples can be slightly faster than lists for iteration.
- **Dictionary Keys:** Tuples can be used as keys in dictionaries (because they are immutable), while lists cannot.


## 6. Wrap-Up
You have learned how to create and use tuples in Python. Remember that tuples are **immutable** sequences, making them suitable for fixed collections of items.


## 7. Hands-on Tasks

### Task 1: The "Mutable" Tuple Trick
1. Create a tuple `colors` with values `('red', 'green', 'blue')`.
2. You want to add `'yellow'` to it, but tuples are immutable.
3. Convert `colors` to a list.
4. Append `'yellow'` to the list.
5. Convert the list back to a tuple named `colors_v2`.
6. Print `colors_v2`.

### Task 2: User Input & Membership
1. Create a tuple `allowed_ids` with values `(1001, 1002, 1003)`.
2. Ask the user to input their ID code (convert it to an integer).
3. Check if the user's ID is in the `allowed_ids` tuple.
4. Print a message like: `"Access Granted: True"` (or False).


```python
# Task 1 Solution
colors = ('red', 'green', 'blue')
print("Original tuple:", colors)

# Convert to list
temp_list = list(colors)

# Modify list
temp_list.append('yellow')

# Convert back to tuple
colors_v2 = tuple(temp_list)
print("New tuple:", colors_v2)
```

```python
# Task 2 Solution
allowed_ids = (1001, 1002, 1003)

# We use int() from Notebook 1 and input() from Notebook 3
user_id = int(input("Enter your ID code: "))

# We use 'in' which works for tuples just like lists (Notebook 4)
is_allowed = user_id in allowed_ids
print("Access Granted:", is_allowed)
```
