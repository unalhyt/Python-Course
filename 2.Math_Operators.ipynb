## 2. Basic Arithmetic
Python supports standard arithmetic operators like `+` and `-`.
When mixing integers and floats, Python automatically promotes the result to a **float**.

```python
# Basic subtraction
print(x-6)
# Mixed type arithmetic (int + float - int) results in a float
c = x + y - 2
print(c)
```

## 3. Multiplication and Division
- Use `*` for multiplication.
- Use `/` for division. Note that in Python 3, division **always** returns a float, even if the result is a whole number.

```python
# floating-point algebra vs 'real' algebra

# Multiplication
print(x*y)

# to divide use /
# Division always returns a float in Python 3
4/3
```

## 4. Exponentiation (Powers)
Python uses `**` for exponentiation (not `^`).

```python
# power

# 3 to the power of 2
3**2
```

### Operator Precedence Warning
Be careful with order of operations!

```python
# square root without function
# Note: This evaluates as (9**1) / 2 due to precedence. Result is 4.5, not 3.
9**1/2
```

The error here comes from the operation precedence. Instead, you should write 9**(1/2)

## 5. String Operations
Some math operators work on strings too, but they behave differently:
- `+` concatenates (joins) strings.
- `*` repeats a string.

```python
# math with strings

firstName = "Ünal"
lastName = "Hayta"
# String concatenation
print(firstName + ' ' + lastName)
```

```python
# This will raise a TypeError because subtraction is not defined for strings
firstName - lastName
```

```python
# String repetition
firstName * 4
```
