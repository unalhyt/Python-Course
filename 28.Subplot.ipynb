## Python Basics — Subplots

In this notebook, we will explore **subplots**, which allow you to display multiple plots in a single figure. This is essential for comparing different datasets or visualizing multiple aspects of your data side-by-side.

## Learning Goals
By the end of this notebook, students will be able to:
- Create subplots using `plt.subplot()`.
- Create subplots using `plt.subplots()` (object-oriented approach).
- Understand the grid layout (rows, columns).
- Access individual axes objects to customize each plot.
- Use loops to populate a grid of subplots efficiently.


## 1. Using `plt.subplot()`
The `plt.subplot(rows, cols, index)` function activates a specific subplot within a grid. The index starts at 1.

**Example:**

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure with 1 row and 2 columns
plt.figure(figsize=(10, 4))

# First subplot (left)
plt.subplot(1, 2, 1)
plt.plot(np.random.randn(10))
plt.title("Plot 1")

# Second subplot (right)
plt.subplot(1, 2, 2)
plt.plot(np.random.randn(10))
plt.title("Plot 2")

plt.show()
```

## 2. Looping with `plt.subplot()`
You can use a loop to create a grid of plots.


```python
plt.figure(figsize=(8, 8))

for i in range(9):
    # Create a 3x3 grid, activate the (i+1)-th subplot
    plt.subplot(3, 3, i + 1)
    plt.plot(np.random.randn(10))
    plt.title(f"Plot {i+1}")

plt.tight_layout()  # Adjust spacing
plt.show()
```

## 3. Object-Oriented Approach: `plt.subplots()`
The `plt.subplots()` function returns a figure object (`fig`) and an array of axes objects (`ax`). This is often more flexible and cleaner.

**Example (1D Array of Axes):**

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))

x = np.arange(10)

# Access axes by index
ax[0].plot(x, x**2, 'b')
ax[0].set_title("Square")

ax[1].plot(x, np.sqrt(x), 'r')
ax[1].set_title("Square Root")

ax[2].plot(x, x**3, 'g')
ax[2].set_title("Cube")

plt.show()
```

## 4. 2D Array of Axes
When creating a grid with multiple rows and columns, `ax` becomes a 2D array. You access axes using `ax[row, col]`.

**Example (2D Array of Axes):**

```python
fig, ax = plt.subplots(2, 2, figsize=(8, 8))

print("Shape of ax array:", ax.shape)

# Plot on specific subplots
ax[0, 0].plot(np.random.randn(10))
ax[0, 0].set_title("Top Left")

ax[0, 1].plot(np.random.randn(10))
ax[0, 1].set_title("Top Right")

ax[1, 0].plot(np.random.randn(10))
ax[1, 0].set_title("Bottom Left")

ax[1, 1].plot(np.random.randn(10))
ax[1, 1].set_title("Bottom Right")

plt.tight_layout()
plt.show()
```

## 5. Practice Task: Looping over Axes
**Exercise:**
1.  Create a 3x3 grid of subplots using `plt.subplots()`.
2.  Populate the subplots using a `for` loop.
3.  Hint: Use `.flatten()` on the `ax` array to iterate over it as a 1D sequence.


```python
# Create a 3x3 grid
fig, ax = plt.subplots(3, 3, figsize=(10, 10))

# Flatten the 2D array of axes into a 1D array
axes_flat = ax.flatten()

# Loop through each axis and plot something
for i, axis in enumerate(axes_flat):
    axis.plot(np.random.randn(10))
    axis.set_title(f"Subplot {i+1}")

plt.tight_layout()
plt.show()
```
