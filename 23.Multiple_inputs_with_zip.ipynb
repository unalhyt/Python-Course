## Python Basics — Multiple Inputs with Zip

In this notebook, we will explore the `zip()` function, which allows you to iterate over multiple sequences (like lists) simultaneously. This is a powerful tool for combining data from different sources.

## Learning Goals
By the end of this notebook, students will be able to:
- Use `zip()` to combine two or more lists into an iterator of tuples.
- Iterate over multiple lists in a `for` loop using `zip()`.
- Convert a `zip` object into a list of tuples.
- Create a dictionary from two lists using `zip()`.


## 1. Iterating Over Multiple Lists
Suppose you have two lists and you want to process their elements in pairs.

**Example:**

```python
somelist = [3, 4, 5, 6, 7]
otherlist = ['d', 'e', 's', 'm', 'x']

# Traditional way using range and index
print("Using range:")
for i in range(len(somelist)):
    print(otherlist[i] + ' ' + str(somelist[i]))
```

## 2. Using `zip()`
The `zip()` function pairs elements from each list, creating an iterator of tuples. This is cleaner and more Pythonic.


```python
print("\nUsing zip:")
for letter, number in zip(otherlist, somelist):
    print(letter, number)
```

## 3. Zip Objects
The `zip()` function returns a zip object, which is an iterator. To see the contents, you can convert it to a list.

**Note:** Iterators are consumed when you iterate over them or convert them. Once consumed, they are empty.


```python
z = zip(otherlist, somelist)
# print(z[0]) # This would raise a TypeError because zip objects are not subscriptable

# Convert to a list to inspect
lz = list(z)
print("\nList of tuples:", lz)

# Accessing elements
print("Second tuple, second element:", lz[1][1])
```

## 4. Zipping Unequal Lists
If the input lists have different lengths, `zip()` stops when the shortest list is exhausted.


```python
a = [2, 3, 4, 5, 6, 7, 8, 4, 3, 4]
b = ['d', '3', 'f']
c = ['1', 2, 3, 4, 6, 8]

print("\nZipping unequal lists:")
for i, j, k in zip(a, b, c):
    print(i, j, k)
# Notice it stops after 3 iterations because list 'b' has only 3 elements.
```

## 5. Practice Task: Creating a Dictionary
**Exercise:**
Use the `zip()` function to create a dictionary where `names` are keys and `values` are values.


```python
names = ['alpha', 'beta', 'gamma']
values = [10, 20, 40]

# Method 1: Using a loop and update
D1 = {}
for name, val in zip(names, values):
    D1.update({name: val})
print("Dictionary (Method 1):", D1)

# Method 2: Direct conversion (More Pythonic)
D2 = dict(zip(names, values))
print("Dictionary (Method 2):", D2)

# Access a value
print("Value for gamma:", D2['gamma'])
```
